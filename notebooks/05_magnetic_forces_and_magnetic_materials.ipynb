# Magnetic Force and Magnetic Materials

How do magnetic fields exert forces and torques on currents, and how do magnetic materials modify the magnetic field through magnetization?

## Magnetic Force
### Lorentz Force on a Moving Charge

$$\mathbf{F} = q(\mathbf{E} + \mathbf{v} \times \mathbf{B})$$
$$\mathbf{F}_B = q(\mathbf{v} \times \mathbf{B})$$
$$F_B = |q|vB\sin\theta$$

### Force on a Current-Carrying Wire

For on a current element,
$$\mathbf{J} = \rho _v \mathbf{v}$$
$$I d\vec{l} = \frac{dQ}{dt}d\vec{l} = dQ\frac{d\vec{l}}{dt}$$
Therefore,
$$I d\vec{l} = dQ\mathbf{v}$$
Applying this to force,
$$d\mathbf{F} = dQ(\mathbf{v} \times \mathbf{B}) = Id\vec{l} \times \mathbf{B}$$
$$\mathbf{F} = \int I d\vec{l} \times \mathbf{B}$$
also,
$$I d\vec{L} = \mathbf{k}dS = \mathbf{J}dV,$$
so we can write:
$$\mathbf{F} = \iint \mathbf{K}\,ds  \times \mathbf{B}, \quad \mathbf{F} = \iiint \mathbf{J}\,dV  \times \mathbf{B}$$

### Force Between Parallel Currents

If you use the Biot-Savart law,
$$d\mathbf{B}_1(\mathbf{r}_2) = \frac{\mu _0 I_1}{2\pi d}\hat{\phi}$$
$$\frac{\mathbf{F}_2}{L} = I_2 \hat{\mathbf{u}_2} \times \mathbf{B}_1(\mathbf{r}_2)$$
Therefore,
$$\frac{F_2}{L} = \frac{\mu _0 \left| I_1 I_2 \right|}{2\pi d}$$


## Torque and Magnetic Dipoles
### Torque on a Current Loop
No rotational force is generated when the magnetic field is perpendicular to the current loop, but a rotational force is generated when the magnetic field and the current loop are parallel. Expressed mathematically, this is as follows:

$$\boldsymbol{\tau} = \boldsymbol{\mu} \times \mathbf{B}, \quad \boldsymbol{\mu} = IA\hat{n}$$

$\boldsymbol{\mu}$ is the magnetic dipole moment, and for a coil wound N times, it can be written as follows.
$$\boldsymbol{\mu} = NIA \hat{n}$$

The direction of $\hat{n}$ is determined using the right-hand rule, and the magnetic field attempts to align $\mu$ parallel to the magnetic field. 

Because of the torque equation:
$$\boldsymbol{\tau} = \boldsymbol{\mu} \times \mathbf{B}$$

A rotational force is generated when the current loop is oriented horizontally relative to the magnetic field, and the rotation stops when it is oriented vertically.
We can get the magnitude of torque:
$$\tau = \mu B \sin\theta$$

The torque is zero when the magnetic dipole moment is parallel or antiparallel to the magnetic field, and its magnitude is maximum when the two vectors are perpendicular.

## Magnetic Materials
### Magnetization

$$\mathbf{M} = \frac{d\mathbf{\mu}}{dV}$$ 
in $[\mathbf{M}] = \mathrm{A/m}$ 

Magnetization represents the collective magnetic response of microscopic magnetic dipoles inside a material.

### Magnetic Field Intensity

|Symbol|Name|SI unit|
|:---:|:---:|:---:|
|$\mathbf{B}$|magnetic flux density|T|
|$\mathbf{H}$|magnetic field intensity|A/m|
|$\mathbf{M}$|magnetization|A/m|

$$\mathbf{B} = \mu _0 (\mathbf{H} + \mathbf{M})$$

### Susceptibility and Permeability

In **linear magnetic materials**, magnetization is proportional to the magnetic field strength,
$$\mathbf{M} = \chi _m \mathbf{H}$$
$\chi _m$: magnetic susceptibility

Substituting this into the equation below,
$$\mathbf{B} = \mu _0 (\mathbf{H} + \mathbf{M})$$
You can obtain the following.
$$\mathbf{B} = \mu _0(1 + \chi _m) \mathbf{H}$$
Defining the relative permeability ass,
$$\mu _r = 1 + \chi _m$$

the constitutive relation becomes
$$\mathbf{B} = \mu _0 \mu _r \mathbf{H} = \mu\mathbf{H}$$

The relation $B =\mu H$ is a constitutive model for a linear and isotropic magnetic material; it is not a universal relation for all magnetic materials.

### Classification of Magnetic Materials
|Material type|Susceptibility|Response|
|:---:|:---:|:---:|
|Diamagnetic|$\chi _m < 0$|It is weakly magnetized in the opposite direction to the external field.|
|Paramagnetic| $\chi _m > 0$ |It is weakly magnetized in the same direction as the external field. |
|Ferromagnetic| generally nonlinear |It becomes strongly magnetized, and residual magnetism may appear.|

```python
r_1 = np.array([0, 0, 0])
u_1 = np.array([0, 0, 1])
I_1 = 5.0 # [A]

d = 0.10 # [m]

r_2 = np.array([d, 0, 0])
u_2 = np.array([0, 0, 1])
I_2 = 3.0 # [A]

magnetic_field_A = magnetic_field_infinite_straight_wire(r_2, I_1, r_1, u_1)
magnetic_field_magnitude_A = np.linalg.norm(magnetic_field_A)
magnetic_field_reference_A = MU_0 * abs(I_1) / (2 * np.pi * d)
R_A = relative_error(magnetic_field_magnitude_A, magnetic_field_reference_A)

F_2_per_L_B = I_2 * np.cross(u_2, magnetic_field_A)
F_2_per_L_magnitude_B = np.linalg.norm(F_2_per_L_B)
F_2_per_L_reference_B = MU_0 * abs(I_1 * I_2) / (2 * np.pi * d)
R_B = relative_error(F_2_per_L_magnitude_B, F_2_per_L_reference_B)

I_2_opposite = -I_2
F_2_per_L_C = I_2_opposite * np.cross(u_2, magnetic_field_A)
F_2_per_L_magnitude_C = np.linalg.norm(F_2_per_L_C)
```

```python
d_range = np.linspace(0.01, 0.50, 300)
F_L_D = []

for d in d_range:
    r_2_D = np.array([d, 0, 0])
    F_2_per_L_D = I_2 * np.cross(u_2, magnetic_field_infinite_straight_wire(r_2_D, I_1, r_1, u_1))
    F_L_D.append(F_2_per_L_D)

F_L_D = np.asarray(F_L_D) 
F_L_D_magnitude = np.linalg.norm(F_L_D, axis=1)

F_L_D_reference = MU_0 * abs(I_1 * I_2) / (2 * np.pi * d_range)
F_L_D_reference = np.asarray(F_L_D_reference)

plt.figure(figsize=(6, 4))

plt.plot(
    d_range,
    F_L_D_magnitude,
    label="F/L",
    linewidth=2.2,
    color=COLORS["numerical"]
)

plt.plot(
    d_range,
    F_L_D_reference,
    label="analytical",
    linestyle="-.",
    linewidth=1.8,
    color=COLORS["analytical"]
)

plt.xlabel(r"Wire separation, $d[m]$")
plt.ylabel(r"Force per unit length, $F/L \mathrm{[N/m]}$")
plt.title("Magnetic Force per Unit Length versus Wire Separation")

plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()

```

```python
F_L_E = []
for d in d_range:
    r_2_E = np.array([d, 0, 0])
    F_2_per_L_E = d * I_2 * np.cross(u_2, magnetic_field_infinite_straight_wire(r_2_E, I_1, r_1, u_1))
    F_L_E.append(F_2_per_L_E)

F_L_E = np.asarray(F_L_E) 
F_L_E_magnitude = np.linalg.norm(F_L_E, axis=1)

analytical_value = MU_0 * abs(I_1 * I_2) / (2 * np.pi)

plt.figure(figsize=(6, 4))

plt.plot(
    d_range,
    F_L_E_magnitude,
    label=r"d x F/L",
    linewidth=2.2,
    color=COLORS["numerical"]
)

plt.axhline(
    analytical_value,
    color=COLORS["reference"],
    linestyle="--",
    linewidth=1.2,
    label="analytical_value"
)

plt.xlabel(r"Wire separation, $d[m]$")
plt.ylabel(r"Force per unit length, $d \times F/L \mathrm{[N/m]}$")
plt.title("Magnetic Force per Unit Length versus Wire Separation")

plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()
```
