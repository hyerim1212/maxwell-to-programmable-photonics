# Wave Superposition, Polarization and Interference

How do the relative amplitude and phase of electromagnetic-wave components determine interference and polarization, and how can these effects be represented using complex vectors and matrices?

$$\mathbf{E}(\mathbf{r},t)=\text{Re}\left\{\mathbf{E}_0e^{i\alpha(\mathbf{r})}e^{-i\omega t}\right\}$$

Handedness is defined with respect to the propagation direction.

## 1. Superposition of Harmonic Waves

### 1.1 Principle of Superposition

Each component of the electromagnetic wave $E_x, E_y, E_z, B_x, B_y, B_z$ satisfies the 3 - dimension scalar wave equation, 
$$\frac{\partial^2 \psi}{\partial x^2} + \frac{\partial^2 \psi}{\partial y^2} +\frac{\partial^2 \psi}{\partial z^2} = \frac{1}{v^2}\frac{\partial^2 \psi}{\partial t^2}$$

That equation is linear, so $\psi (\vec{\mathbf{r}}, t)$and the difference value has the structure of a 1-order term.  

If $\psi_1 (\vec{\mathbf{r}}, t), \cdots, \psi_n (\vec{\mathbf{r}}, t)$ is a solution of that equation, Their linear combination is also a solution.

Therfore, 
$$\psi (\vec{\mathbf{r}}, t) = \sum_{i = 1}^{n}C_i \psi_i(\vec{\mathbf{r}}, t)$$ means that wave equation, and $C_i$ is an arbitrary constant.

### 1.2 Complex Amplitude Representation
when given a wave,
$$E_1 = E_{01}\cos(kx \pm \omega t + \varepsilon _1)$$

We can write that like this:
$$E_1 = E_{01}\cos(\alpha _1 \mp \omega t )$$

and when we use a complex form,
$$\tilde{E}(\mathbf{r}) = E_0\,e^{i\alpha(\mathbf{r})} $$

That is the same as
$$E_0\cos(\alpha \pm \omega t)$$

and we can write the equation with the sum of the component waves, 
$$\tilde{E} = \sum_{j = 1}^{N} E_{0j}\, e^{i\alpha _j}$$

$$\mathbf{E}(\mathbf{r},t)=\text{Re}\left\{\mathbf{E}_0e^{i\alpha(\mathbf{r})}e^{-i\omega t}\right\}$$

It is the complex amplitude of the composite wave, which is equivalent to the simplicity of the complex amplitude of individual waves.
$$E_0^2 = (E_0e^{i\alpha})(E_0e^{-i\alpha})$$

If, $N = 2$

$$E_0 ^2 = (E_{01}e^{i\alpha _1} + E_{02} e^{i\alpha _2})(E_{01}e^{-i\alpha _1} + E_{02} e^{-i\alpha _2})$$
So,
$$E_0 ^2 = E_{01}^2 + E_{02}^2 + 2E_{01}E_{02}\cos(\alpha_1 - \alpha_2)$$

### 1.3 Relative Phase

$$\Delta\varphi = \varphi_2 - \varphi_1$$

$\Delta\varphi = 0$ : in phase

$\Delta\varphi = \pi$ : out of phase

## 2. Two-Wave Interference

### 2.1 Interference Intensity

I regarded intensity in `07_electromagnetic_waves_and_dielectric_interfaces`
$$\mathbf{E} = \mathbf{E}_1 + \mathbf{E}_2 + \cdots $$ 

when 
$$\mathbf{E} = \mathbf{E}_1 + \mathbf{E}_2 \quad \text{in line-polarization},$$

we can wirte intensity:
$$I = \epsilon v \left< \mathbf{E}^2 \right>_T$$

and also,
$$\mathbf{E}^2 = \mathbf{E} \cdot \mathbf{E} = (\mathbf{E}_1 + \mathbf{E}_2)(\mathbf{E}_1 + \mathbf{E}_2)$$

$$\mathbf{E}^2 = \mathbf{E}_1^2 + \mathbf{E}_2^2 +2 \mathbf{E}_1\cdot\mathbf{E}_2$$

Furthermore, we can describe the electric field with intensity:
$$I = I_1 + I_2 + I_{12}$$

Let's use the time average value,
$$I_1 = \epsilon v\left< E_1^2 \right>_T,\quad I_2 = \epsilon v\left< E_2^2 \right>_T, \quad I_{12} = 2\epsilon v\left< \mathbf{E}_1\cdot\mathbf{E}_2 \right>_T$$

$I_{12} = 2\epsilon v\left< \mathbf{E}_1\cdot\mathbf{E}_2 \right>_T$ is interference term.

$$\mathbf{E}_1 \cdot \mathbf{E}_2 = \mathbf{E}_{01} \cdot \mathbf{E}_{02}\cos(\alpha_1 - \omega t) \times \cos(\alpha_2 - \omega t) $$
$$\left<\mathbf{E}_1 \cdot \mathbf{E}_2 \right>_T = \frac{1}{2} \mathbf{E}_{01} \cdot \mathbf{E}_{02}\cos(\mathbf{k}_1\cdot\mathbf{r}-\mathbf{k}_2\cdot\mathbf{r} + \boldsymbol{\varepsilon} _1 - \boldsymbol{\varepsilon} _2)$$

Because of
$$\left<\cos ^2 \omega t \right>_T = \left<\sin ^2 \omega t \right>_T = \frac{1}{2}$$
$$\left<\cos \omega t \, \sin \omega t \right>_T = 0,$$

Therefore,
$$I_{12} = \epsilon v\mathbf{E}_{01} \cdot \mathbf{E}_{02} \cos\Delta\varphi, \quad \Delta\varphi = (\mathbf{k_1}\cdot\mathbf{r}-\mathbf{k}_2\cdot\mathbf{r} + \boldsymbol{\varepsilon} _1 - \boldsymbol{\varepsilon} _2)$$

To write the interference term simply,
$$I_1 = \epsilon v\left< {E}_1^2 \right>_T = \frac{\epsilon v E_{01} ^2}{2}, \quad I_2 = \epsilon v \left< {E}_2^2 \right>_T = \frac{\epsilon v E_{02} ^2}{2}$$
$$I = I_1 + I_2 + 2\sqrt{I_1I_2}\cos\Delta\varphi$$

### 2.2 Constructive and Destructive Interference

We can call it total constructive interference when:
$$I_\text{max}: I_1 + I_2 + 2\sqrt{I_1I_2} \quad \text{in} \,\, \Delta\varphi = 2k\pi, \quad k \in \mathbb{Z}$$

When $I_1 = I_2,$ we can call it total destructive interference when:
$$I_\text{min}: I_1 + I_2 - 2\sqrt{I_1I_2} \quad \text{in} \,\, \Delta\varphi = (2\ell-1)\pi, \quad \ell \in \mathbb{Z}$$


constructive: $I_1 + I_2 < I < I_\text{max}$

destructive: $I_1 + I_2 > I > I_\text{min}$

### 2.3 Interference Visibility

$$V = \frac{I_\text{max} - I_\text{min}}{I_\text{max} + I_\text{min}}$$
$$V = \frac{2\sqrt{I_1I_2}}{I_1+I_2}$$

## 3. Polarization as Vector Superposition

### 3.1 Orthogonal Field Components

$$
\tilde{\mathbf{E}} = \begin{bmatrix}
E_0x e^{i\varphi_x} \\
E_0y e^{i\varphi_y}  
\end{bmatrix}
$$
$$\Delta\varphi = \varphi_y - \varphi_x$$

### 3.2 Linear Polarization

when $\Delta\varphi = 2n\pi, \quad n \in \mathbb{Z}$
$$\mathbf{E} = (\hat{i}E_{0x} + \hat{j}E_{0y})\cos(kz - \omega t)$$

The vector $(\hat{i}E_{0x} + \hat{j}E_{0y})$ is constant, so the electric field oscillates along a fixed direction. 

Therefore, the wave is linearly polarized.

$$\tan\theta = \frac{E_{0y}}{E_{0x}}$$
The electric field is oscillating with $\theta$. 

when $\Delta\varphi = (2n-1)\pi, \quad n \in \mathbb{Z}$
$$\mathbf{E} = (\hat{i}E_{0x} - \hat{j}E_{0y})\cos(kz - \omega t)$$

The vector $(\hat{i}E_{0x} - \hat{j}E_{0y})$ is constant, so the electric field oscillates along a fixed direction. 

Some linearly polarized waves can decompose into two perpendicular components.


### 3.3 Circular Polarization

When $E_{0x} = E_{0y},$

when $\Delta\varphi = -\frac{\pi}{2} + 2n\pi, \quad n \in \mathbb{Z}$

$$\mathbf{E} = E_0\left[ \hat{i}\cos(kz - \omega t) + \hat{j}\sin(kz - \omega t)\right]$$

The electric-field vector rotates at angular frequency $\omega$, and it rotates clockwise.

When $\Delta\varphi = \frac{\pi}{2} + 2n\pi, \quad n \in \mathbb{Z}$, there is no change in amplitude, but it rotates counterclockwise.

We can describe linear and elliptical polarizations in terms of circular polarization.

### 3.4 Elliptical Polarization

$$\left(\frac{E_{x}}{E_{0x}}\right)^2 + \left(\frac{E_{y}}{E_{0y}}\right)^2 -2\left(\frac{E_{x}}{E_{0x}}\right)\left(\frac{E_{y}}{E_{0y}}\right)\cos\Delta\varphi = \sin^2\Delta\varphi$$
It is an elliptical equation has angulat $\alpha$ on the $E_x - E_y$ coordinate system,
$$\tan2\alpha = \frac{2E_{0x}E_{0y}\cos\Delta\varphi}{E_{0x}^2-E_{0y}^2}$$

## Case 1: Two-Wave Interference and Visibility

### Physical Question

How do relative phase and amplitude imbalance between two coherent waves determine the total intensity and interference visibility?

### Model and Assumptions

Consider two coherent, monochromatic, linearly polarized waves propagating in the same direction and having the same polarization.

Their complex field amplitudes are
$$\tilde{E}_1 = A_1e^{i\varphi_1}, \quad \tilde{E}_2 = A_2e^{i\varphi_2}$$

with relative phase
$$\Delta\varphi = \varphi_2 - \varphi_1$$

Assume that intensity is represented in normalized form by
$$I = \left|\tilde{E}\right|^2$$

The analytical interference relation is
$$I = I_1 + I_2 + 2\sqrt{I_1I_2}\cos\Delta\varphi$$

### Task A — Phase-Dependent Interference

Let
$$A_1 = A_2 = 1$$

Sweep the relative phase over
$$0 \leq \Delta\varphi \leq 2\pi$$

#### Verification A

Quantitatively verify that
$$I(0) = 4$$
$$I(\pi) = 0$$
$$I(\frac{\pi}{2}) = 2$$

### Task B — Visibility under Amplitude Imbalance
Fix

$$A_1 = 1$$

and vary the second-wave amplitude.

For each amplitude ratio
$$r = \frac{A_2}{A_1}$$

determine the maximum and minimum interference intensities from a phase sweep and calculate
$$V_\text{analytical} = \frac{I_\text{max} - I_\text{min}}{I_\text{max} + I_\text{min}}$$

Compare the result with
$$V_\text{analytical} = \frac{2A_1A_2}{A_1^2 + A_2^2}$$

Plot visibility versus amplitude ratio.

#### Verification B

Use several representative amplitude ratios, for example,
$$r = 1, \quad 0.75, \quad 0.5, \quad 0.25$$

and quantitatively compare numerical and analytical visibility.

### Physical Interpretation

Answer the following questions.

In this case, intensity is represented by $|\tilde E|^2$, so the plotted quantity is proportional to physical optical intensity rather than expressed in SI units.

```python
A_1 = 1.0
A_2 = 1.0

delta_phi_range = np.linspace(0.0, 2.0 * np.pi, 201)
```

```python
field_1 = complex_field(A_1, 0.0)
field_2 = complex_field(A_2, delta_phi_range)

field_total = superpose_fields(field_1, field_2)

I_entire = field_intensity(field_total)

I_0 = I_entire[0]
I_pi_over_2 = I_entire[50]
I_pi = I_entire[100]
I_2pi = I_entire[200]

I_analytical = (A_1**2 + A_2**2 + 2.0 * A_1 * A_2 * np.cos(delta_phi_range))
intensity_error = absolute_error(I_entire, I_analytical)

print("I(0):", I_0)
print("I(pi/2):", I_pi_over_2)
print("I(pi):", I_pi)
print("I(2pi):", I_2pi)

print("Maximum analytical comparison error:", np.max(intensity_error))
```

```python
r_range = np.linspace(0.0, 1, 201)
r_spot = np.array([1, 0.75, 0.5, 0.25])

V_spot = 2.0 * r_spot / (1.0 + r_spot**2)

V_numerical = []

for r in r_range:
    A_2 = r * A_1

    field_2 = complex_field(A_2, delta_phi_range)
    field_total_B = superpose_fields(field_1, field_2)

    I_entire = field_intensity(field_total_B)

    I_max = np.max(I_entire)
    I_min = np.min(I_entire)

    V = (I_max - I_min) / (I_max + I_min)
    V_numerical.append(V)

V_numerical = np.asarray(V_numerical)

V_analytical = 2.0 * r_range / (1.0 + r_range**2)

plt.plot(
    r_range,
    V_numerical,
    label="Numerical",
    color=COLORS["numerical"]
)

plt.plot(
    r_range,
    V_analytical,
    linestyle="--",
    label="Analytical",
    color=COLORS["analytical"]
)

plt.scatter(
    r_spot,
    V_spot,
    label="Selected amplitude ratios",
    color=COLORS["numerical_alt"]
)

plt.xlabel(r"Amplitude ratio $r=A_2/A_1$")
plt.ylabel("Visibility")
plt.title("Interference Visibility vs Amplitude Ratio")

plt.legend()
plt.grid()

plt.show()

```

```python
r_spot = np.asarray([1.0, 0.75, 0.5, 0.25])

V_spot_numerical = []

for r in r_spot:
    A_2 = r * A_1

    field_2 = complex_field(A_2, delta_phi_range)
    field_total = superpose_fields(field_1, field_2)

    intensity = field_intensity(field_total)

    I_max = np.max(intensity)
    I_min = np.min(intensity)

    visibility = ((I_max - I_min) / (I_max + I_min))
    V_spot_numerical.append(visibility)

V_spot_numerical = np.asarray(V_spot_numerical)

V_spot_analytical = (2.0 * r_spot / (1.0 + r_spot**2))

visibility_error = absolute_error(V_spot_numerical, V_spot_analytical)

print("Amplitude ratios:", r_spot)
print("Numerical visibility:", V_spot_numerical)
print("Analytical visibility:", V_spot_analytical)
print("Absolute error:", visibility_error)
print("Maximum error:", np.max(visibility_error))
```

## Case 3: Phase Retardation and Jones-Matrix Transformation

### Physical Question

How does a relative phase delay between orthogonal field components transform the polarization state of light, and how can this transformation be represented by a Jones matrix?

### Model and Assumptions

Consider an ideal lossless phase retarder whose principal axes coincide with the \(x\)- and \(y\)-axes.

Represent the device by

$$
J(\delta) = \begin{bmatrix}
1 & 0\\
0 & e^{i\delta}
\end{bmatrix}
$$

where $\delta$ is the relative phase retardation.

The output polarization state is
$$\tilde{\mathbf{E}}_\text{out} = J(\delta)\tilde{\mathbf{E}}_\text{in}$$

Assume that the device changes phase but introduces no loss.

### Task A — Transformation of a $45^\circ$ Linear Input

Use the normalized input state
$$
\tilde{\mathbf{E}}_\text{in} = \frac{1}{\sqrt{2}}\begin{bmatrix}
1 \\
1
\end{bmatrix}
$$

Calculate the output state for
$$\delta = 0, \quad \delta = \frac{\pi}{2}, \quad \delta = \pi$$

Visualize the corresponding polarization trajectory in the $E_x - E_y$ plane.

Identify each output polarization state.

#### Verification A

Verify quantitatively that the total normalized field magnitude is conserved:
$$\tilde{\mathbf{E}}_\text{out}^\dagger\tilde{\mathbf{E}}_\text{out} = \tilde{\mathbf{E}}_\text{in}^\dagger\tilde{\mathbf{E}}_\text{in}$$

Also verify directly that
$$J^\dagger J = I$$

Interpret this result physically.

### Task B — Continuous Phase Retardation

Sweep
$$0 \le \delta \le 2\pi$$

For the same $45^\circ$ linear input, determine how the polarization state evolves as the phase delay increases.

Select physically important phase delays and classify the corresponding output states.

### Task C — Analyzer Measurement

Pass the output through a linear analyzer with a Jones vector

$$
\mathbf{a}(\theta) = \begin{bmatrix}
\cos\theta \\
\sin\theta
\end{bmatrix}
$$

Calculate
$$\tilde{E}_\text{proj} = \mathbf{a}^\dagger\tilde{\mathbf{E}}_\text{out}$$

and determine the transmitted intensity
$$I(\theta) = \left|\tilde{E}_\text{proj}\right|^2$$

Compare analyzer-angle responses for linear polarization, circular polarization, one elliptical polarization state.

#### Verification C

For a normalized linear polarization state aligned at angle $\theta_0$, verify Malus' law:
$$I(\theta) = \cos^2(\theta - \theta_0)$$

Use numerical error/tolerance functions rather than judging agreement from the graph.

### Physical Interpretation

1. Why does a lossless phase retarder change polarization without changing total optical power?
2. Why does a quarter-wave retardation transform a $45^\circ$ linear state into circular polarization?
3. What transformation is produced by a half-wave retardation?
4. Why is a phase retarder represented by a matrix rather than by adding two fields?
5. What is the difference between the superposition of fields and the linear transformation of a field state?
6. How does this matrix formulation prepare the mathematical framework for optical components such as phase shifters and Mach–Zehnder interferometers?

```python
delta = np.asarray([0, np.pi / 2, np.pi])

E_out = []

E_in = 1 / np.sqrt(2) * np.array([
        [1],
        [1]], dtype = float)

for d in delta :
    
    J = np.array([
        [1.0, 0.0],
        [0.0, np.exp(1j * d)]
    ], dtype=complex)

    E = J @ E_in

    E_out.append(E)

E_out = np.asarray(E_out)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

titles = [
    r"$\delta = 0$",
    r"$\delta = \pi/2$",
    r"$\delta = \pi$"
]

for i, ax in enumerate(axes):

    E_x_complex = E_out[i, 0, 0]
    E_y_complex = E_out[i, 1, 0]

    E_x = np.real(E_x_complex * np.exp(-1j * phase_time))
    E_y = np.real(E_y_complex * np.exp(-1j * phase_time))

    ax.plot(E_x, E_y)

    ax.set_title(titles[i])
    ax.set_xlabel(r"$E_x$")
    ax.set_ylabel(r"$E_y$")
    ax.set_aspect("equal")
    ax.set_xlim(-0.8, 0.8)
    ax.set_ylim(-0.8, 0.8)
    ax.grid()

plt.tight_layout()
plt.show()

unitarity_error = []

for delta_value in delta:
    J = np.diag([
        1.0,
        np.exp(1j * delta_value)
    ])

    error = complex_error(J.conj().T @ J, np.eye(2))

    unitarity_error.append(np.max(error))

unitarity_error = np.asarray(unitarity_error)

print("Maximum unitarity error:", np.max(unitarity_error))
```
