# Wave Superposition, Polarization and Interference

How do the relative amplitude and phase of electromagnetic-wave components determine interference and polarization, and how can these effects be represented using complex vectors and matrices?

## 1. Superposition of Harmonic Waves

### 1.1 Principle of Superposition

Each component of the electromagnetic wave $E_x, E_y, E_z, B_x, B_y, B_z$ satisfies the 3 - dimension scalar wave equation, 
$$\frac{\partial^2 \psi}{\partial x^2} + \frac{\partial^2 \psi}{\partial y^2} +\frac{\partial^2 \psi}{\partial z^2} = \frac{1}{v^2}\frac{\partial^2 \psi}{\partial t^2}$$

That equation is linear, so $\psi (\vec{\mathbf{r}}, t)$and the difference value has the structure of a 1-order term.  

If $\psi_1 (\vec{\mathbf{r}}, t), \cdots, \psi_n (\vec{\mathbf{r}}, t)$ is a solution of that equation, Their linear combination is also a solution.

Therfore, 
$$\psi (\vec{\mathbf{r}}, t) = \sum_{i = 1}^{n}C_i \psi_i(\vec{\mathbf{r}}, t)$$ means that wave equation, and $C_i$ is an arbitrary constant.

### 1.2 Complex Amplitude Representation
when given wave,
$$E_1 = E_{01}\cos(kx \pm \omega t + \varepsilon _1)$$

we can wirte that like this:
$$E_1 = E_{01}\cos(\alpha _1 \mp \omega t )$$

and when we use complex form,
$$\tilde{E} = E_0\,e^{i(\alpha _1 \mp \omega t)} $$

That is same with
$$E_0\cos(\alpha \pm \omega t)$$

and we can wirte the equation with sum of component wave, 
$$\tilde{E} = \left[ \sum_{j = 1}^{N} E_{0j}\, e^{i\alpha _j}\right]e^{i\omega t}$$

 It is  complex amltitude of the composite wave, this is equivalent to the simplicity of the complex amplitude of individual waves.
$$E_0^2 = (E_0e^{i\alpha})(E_0e^{i\alpha})^\ast$$

If, $N = 2$

$$E_0 ^2 = (E_{01}e^{i\alpha _1} + E_{02} e^{i\alpha _2})(E_{01}e^{-i\alpha _1} + E_{02} e^{-i\alpha _2})$$
So,
$$E_0 ^2 = E_{01}^2 + E_{02}^2 + 2E_{01}E_{02}\cos(\alpha_1 - \alpha_2)$$

## 2. Two-Wave Interference

### 2.1 Interference Intensity

I regarded intensity in `07_electromagnetic_waves_and_dielectric_interfaces`
$$\mathbf{E} = \mathbf{E}_1 + \mathbf{E}_2 + \cdots $$ 
when $$\mathbf{E} = \mathbf{E}_1 + \mathbf{E}_2 \quad \text{in p-state}$$,
$$I = \epsilon v \left< \mathbf{E}^2 \right>_T$$
$$\mathbf{E}^2 = \mathbf{E} \cdot \mathbf{E} = (\mathbf{E}_1 + \mathbf{E}_2)(\mathbf{E}_1 + \mathbf{E}_2)$$

$$\mathbf{E}^2 = \mathbf{E}_1^2 + \mathbf{E}_2^2 +2 \mathbf{E}_1\cdot\mathbf{E}_2$$
$$I = I_1 + I_2 + I_{12}$$

$$I_1 = \left< \mathbf{E}_1^2 \right>_T,\quad I_2 = \left< \mathbf{E}_2^2 \right>_T, \quad I_{12} = 2\left< \mathbf{E}_1\cdot\mathbf{E}_2 \right>_T$$
$I_{12} = 2\left< \mathbf{E}_1\cdot\mathbf{E}_2 \right>_T$ is interference term.
$$\mathbf{E}_1 \cdot \mathbf{E}_2 = E_{01}e^{i\phi_1}\,E_{02}e^{i\phi_2} = \mathbf{E}_{01} \cdot \mathbf{E}_{02}\cos(\alpha_1 - \omega t) \times \cos(\alpha_2 - \omega t) $$
$$\left<\mathbf{E}_1 \cdot \mathbf{E}_2 \right>_T = \frac{1}{2} \mathbf{E}_{01} \cdot \mathbf{E}_{02}\cos(\mathbf{k}\cdot\mathbf{r}-\mathbf{k}\cdot\mathbf{r} + \boldsymbol{\varepsilon} _1 - \boldsymbol{\varepsilon} _2)$$

$$\left<\cos ^2 \omega t \right>_T = \left<\sin ^2 \omega t \right>_T = \frac{1}{2}$$
$$\left<\cos \omega t \, \sin \omega t \right>_T = 0$$
Therefore,
$$I_{12} = \mathbf{E}_{01} \cdot \mathbf{E}_{02} \cos\Delta\phi, \quad \Delta\phi = (\mathbf{k}\cdot\mathbf{r}-\mathbf{k}\cdot\mathbf{r} + \boldsymbol{\varepsilon} _1 - \boldsymbol{\varepsilon} _2)$$

To write the interference term simply,
$$I_1 = \left< \mathbf{E}_1^2 \right>_T = \frac{E_{01} ^2}{2}, \quad I_2 = \left< \mathbf{E}_2^2 \right>_T = \frac{E_{02} ^2}{2}$$
$$I = I_1 + I_2 + 2\sqrt{I_1I_2}\cos\Delta\phi$$

### 2.2 Constructive and Destructive Interference

we can call total constructive interference when:
$$I_\text{max}: I_1 + I_2 + 2\sqrt{I_1I_2} \quad \text{in} \,\, \Delta\phi = 2k\pi, \quad k \in \mathbb{Z}$$

We can call total destructive interference when:
$$I_\text{max}: I_1 + I_2 - 2\sqrt{I_1I_2} \quad \text{in} \,\, \Delta\phi = (2\ell-1)\pi, \quad \ell \in \mathbb{Z}$$


constructive: $I_1 + I_2 < I < I_\text{max}$

destructive: $I_1 + I_2 > I > I_\text{min}$

### 2.3 Interference Visibility

$$V = \frac{I_\text{max} - I_\text{min}}{I_\text{max} + I_\text{min}}$$

### 2.4 Optical Path and Phase Delay
$$\Delta\phi = \frac{2\pi}{\lambda _0}n\Delta L$$

## 3. Polarization as Vector Superposition

### 3.1 Orthogonal Field Components

$$
\tilde{\mathbf{E}} = \begin{bmatrix}
E_0x e^{i\varphi_x} \\
E_0y e^{i\varphi_y}  
\end{bmatrix}
$$
$$\Delta\phi = \phi_y - \phi_x$$

### 3.2 Linear Polarization

**p - state**

when $\Delta\phi = 2n\pi, \quad n \in \mathbb{Z}$
$$\mathbf{E} = (\hat{i}E_{0x} + \hat{j}E_{0y})\cos(kz - \omega t)$$
It has constant amplitude. It has linear polarization.

$$\tan\theta = \frac{E_{0y}}{E_{0x}}$$
The electric field is oscillating with $\theta$. 

Some linearly polarized waves can decompose into two perpendicular components.


### 3.3 Circular Polarization

**R - state**

when $\Delta\phi = -\frac{\pi}{2} + 2n\pi, \quad n \in \mathbb{Z}$

$$\mathbf{E} = E_0\left[ \hat{i}\cos(kz - \omega t) + \hat{j}\sin(kz - \omega t)\right]$$
$$\mathbf{E}_x = \hat{i}\cos (kz_0), \quad \mathbf{E}_y = \hat{j}\sin (kz_0)$$

The electric field vector $\mathbf{E}$ shows to the observer angular frequency $\omega$, and it rotates clockwise.

**L - state**

When $\Delta\phi = \frac{\pi}{2} + 2n\pi, \quad n \in \mathbb{Z}$, there is no change in amplitude, but it rotates counterclockwise.

We can describe the p-state and e-state in terms of the L-state and the R-state.

### 3.4 Elliptical Polarization
**e - state**
$$\left(\frac{E_{y}}{E_{0y}}\right)^2 + \left(\frac{E_{y}}{E_{0y}}\right)^2 -2\left(\frac{E_{y}}{E_{0y}}\right)\left(\frac{E_{y}}{E_{0y}}\right)\cos\Delta\phi = \sin^2\Delta\phi$$
It is an elliptical equation has angulat $\alpha$ on the $E_x - E_y$ coordinate system,
$$\tan2\alpha = \frac{2E_{0x}E_{0y}\cos\Delta\phi}{E_{0x}^2-E_{0y}^2}$$

## 4. Jones Vector Representation

### 4.1 Polarization Basis

$$
\mathbf{J} = \begin{bmatrix}
E_0x e^{i\varphi_x} \\
E_0y e^{i\varphi_y}  
\end{bmatrix}
$$

$$
\mathbf{J} = \begin{bmatrix}
E_x  \\
E_y  
\end{bmatrix}
$$



### 4.2 Representative Jones Vectors

$$
\mathbf{J}_x = \begin{bmatrix}
1 \\
0  
\end{bmatrix},
\qquad
\mathbf{J}_y = \begin{bmatrix}
0 \\
1  
\end{bmatrix}
$$

$$
\mathbf{J}_{45^\circ} = \frac{1}{\sqrt{2}}\begin{bmatrix}
1 \\
1  
\end{bmatrix}
\qquad
\mathbf{J}_{-45^\circ} = \frac{1}{\sqrt{2}}\begin{bmatrix}
1 \\
-1  
\end{bmatrix}
$$

$$
\mathbf{J}_{R} = \frac{1}{\sqrt{2}}\begin{bmatrix}
1 \\
-i  
\end{bmatrix}
\qquad
\mathbf{J}_{L} = \frac{1}{\sqrt{2}}\begin{bmatrix}
1 \\
i  
\end{bmatrix}
$$

### 4.3 Global and Relative Phase

- Young's Double Slit

Global phase is canceled out:
$$\Delta\phi = (\phi_1 + \phi_\text{global})-(\phi_1 + \phi_\text{global}) = \phi_1 - \phi_2$$

## 5. Jones Matrices

### 5.1 Linear Optical Transformations

When we write a p-state, x-polarization by the Jones vector,

$$
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

If this is made to pass through parallel polarization,

$$
\begin{bmatrix}
1\\
0
\end{bmatrix}
\to
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

and also we write another p-stae, y-polarization by the Jones vector,

$$
\begin{bmatrix}
0\\
1
\end{bmatrix}
$$

If this is made to pass through parallel polarization,

$$
\begin{bmatrix}
0\\
1
\end{bmatrix}
\to
\begin{bmatrix}
0\\
0
\end{bmatrix}
$$

If we write this as a matrix,
$$
P_x = \begin{bmatrix}
1 & 0\\
0 & 0
\end{bmatrix}
$$

$P_x$ makes the input wave to get the x-axis polarization.

### 5.3 Rotation of the Polarization Basis

Let's use new coordinate system with $\theta$
$$
\begin{bmatrix}
1\\
0
\end{bmatrix}
\to
\begin{bmatrix}
\cos\theta\\
-\sin\theta
\end{bmatrix}
$$

$$
\begin{bmatrix}
0\\
1
\end{bmatrix}
\to
\begin{bmatrix}
\sin\theta\\
\cos\theta
\end{bmatrix}
$$

and then, we can write a $R$
$$
R(-\theta) = \begin{bmatrix}
\cos\theta & \sin\theta \\
-\sin\theta & \cos\theta
\end{bmatrix}
$$
and also,
$$
R(\theta) = \begin{bmatrix}
\cos\theta & -\sin\theta \\
\sin\theta & \cos\theta
\end{bmatrix}
$$

we can conjugation with $R(\theta)$, $R(-\theta)$:
$$R(\theta)XR(-\theta)$$
$X$ can be $P$, $\omega$, $B$, etc

For example,
$$ P_\theta = R(\theta)PR(-\theta)$$
$$
P_\theta = \begin{bmatrix}
\cos^2\theta & \sin\theta\cos\theta \\
\sin\theta\cos\theta & \sin^2\theta
\end{bmatrix}
$$
we can keep the $P$ 's trait, and we can rotate the $P$ by $\theta$
