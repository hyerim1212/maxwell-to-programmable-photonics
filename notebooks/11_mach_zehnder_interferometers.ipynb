# Mach–Zehnder Interferometers and Phase-Controlled Switching

How does a Mach–Zehnder interferometer achieve phase-controlled optical switching, and how do coupling imperfections affect its switching performance?

## Mach–Zehnder Interferometer: Architecture and Operating Principle

How can a relative phase difference between two optical paths control the output power distribution?

When light is coupled into one of the inputs, it appears to split at the first coupler; the light then accumulates different phases as it travels along the two arms, before being recombined at the second coupler.

## Directional Couplers as Linear Optical Operators

Recall `10_directional_couplers_and_coupled_mode_theory`

$$\frac{d\mathrm{a}}{dz} = -iH\mathrm{a}$$

$$e^{-iHz} = e^{-i\beta z}\begin{bmatrix}
\cos(\kappa z) & -i\sin(\kappa z)\\
-i\sin(\kappa z) & \cos(\kappa z)
\end{bmatrix}$$

Let $L$ be the length of the coupling region,
$$\theta = \kappa L$$

For the same lossless waveguide, excluding the common propagation phase
$$C(\theta) = \begin{bmatrix}
\cos\theta & -i\sin\theta \\
-i\sin\theta & \cos\theta
\end{bmatrix}$$

If we make the light enter the first input,
$$a(0) = \begin{bmatrix}
1\\
0
\end{bmatrix}$$

We can get
$$a(L) = e^{-i\beta L}C(\theta)\begin{bmatrix}
1\\
0
\end{bmatrix}=e^{-i\beta L}\begin{bmatrix}
\cos\theta \\
-i\sin\theta
\end{bmatrix}$$

and also,
$$P_1 = \left| a_\text{out,1} \right|^2, \quad P_2 = \left| a_\text{out,2} \right|^2 $$

### coupling angle

Let's define $\mathbf{b}$:
$$\mathbf{b}(L) = e^{i\beta L}\mathbf{a}(L)$$

We already know the result of the first coupler,
$$\begin{bmatrix}
1\\
0
\end{bmatrix} \rightarrow 
\begin{bmatrix}
\cos\theta\\
-i\sin\theta
\end{bmatrix}$$

Let's
$$x = b_1, \quad y = ib_2$$
$$(x,\, y) = (\cos\theta, \, \sin\theta)$$

So, we can regard $x$, and $y$ to
$$x^2 + y^2 = 1$$
therefore, $(x,\, y)$ describe unit-circle representation of normalized amplitudes of mode. Plus, it is a normalized amplitude representation of specific initial conditions, not a general form.

Also, we defined the propagation constant $\beta_{+}$ and $\beta_{-}$ in `10_directional_couplers_and_coupled_mode_theory`

Furthermore, 
$$\kappa_\text{eff} = \frac{|\beta_{+} - \beta_{-}|}{2}$$

$$\kappa_\text{eff} \approx \kappa_\text{CMT}$$

The coupling angle is one-half of the accumulated relative phase.
$$\theta = \frac{\left|\beta_{+} - \beta_{-}\right|L}{2}$$

Using the $\kappa_{\mathrm{eff}}$ extracted from actual FDM supermode splitting as the coupling coefficient in an ideal CMT framework relies on the assumptions of the two-mode reduced model discussed in `10_directional_couplers_and_coupled_mode_theory`.

### $C_{50:50}$, $50:50$ splitting condition

when $$\theta = \kappa L = \frac{\pi}{4}$$
The two modes have the same output power in the complex amplitude space.

$$C_{50:50} = \frac{1}{\sqrt{2}}\begin{bmatrix}
1 & -i \\
-i & 1
\end{bmatrix}$$

After the end of the coupler, 
$$\begin{bmatrix}
1\\
0
\end{bmatrix} \rightarrow \frac{1}{\sqrt{2}}
\begin{bmatrix}
1\\
-i
\end{bmatrix}$$

It looks like the first coupler splits the light and also the power,
$$P_1 = \cos^2\theta, \quad P_2 = \sin^2\theta$$

therefore,
$$P_1 = P_2 = \frac{1}{2}$$


### MZI

In general, 
$$U_\text{MZI}(\theta_1, \theta_2, \phi) = C(\theta_2)P(\phi)C(\theta_1) $$

So we can get each output to
$$a_\text{out} = \text{Combine} \,@\, \text{Phase} \,@\, \text{Splitting} \,@\, \text{Input}$$

Specially, in $\theta = \frac{\pi}{4}$, MZI usually use this coupler angle. MZI has two couplers and a phase difference,
$$U_\text{MZI} = C_{50:50}P(\phi)C_{50:50}$$

Relative phase $P$ is
$$P(\phi) = \begin{bmatrix}
1 & 0 \\
0 & e^{i\phi}
\end{bmatrix}$$

$\phi$: depends on the relative phase,
$$\phi = \beta_1L_1 - \beta_2L_2$$

When we make the light only into the first input,
$$a_\text{output} = C_{50:50}P(\phi)C_{50:50}\begin{bmatrix}
1\\
0
\end{bmatrix}$$

$$a_\text{output} = \frac{1}{2}\begin{bmatrix}
1-e^{i\phi}\\
-i(1+e^{i\phi})
\end{bmatrix}
$$

and we can get the power, which depends on the relative phase
$$P_1 = \sin^2(\frac{\phi}{2}), \quad P_2 = \cos^2(\frac{\phi}{2})$$

## Unitarity and Power Conservation

### Unitarity and Power Conservation


$$U_\text{MZI}(\theta_1, \theta_2, \phi) = C(\theta_2)P(\phi)C(\theta_1) $$

An ideal coupler and phase shifter are unitary operators,
$$C^\dagger C = I,\quad  P^\dagger P = I$$

So,
$$U_\text{MZI}^\dagger U_\text{MZI} = (C_2PC_1)^\dagger(C_2PC_1) $$
$$U_\text{MZI}^\dagger U_\text{MZI} = C_1^\dagger P^\dagger C_2^\dagger C_2PC_1 $$
$$U_\text{MZI}^\dagger U_\text{MZI} = C_1^\dagger P^\dagger PC_1 $$
$$U_\text{MZI}^\dagger U_\text{MZI} = I $$

Therefore,
$$\|\mathbf{a}_\text{out}\|^2 = \|\mathbf{a}_\text{in}\|^2$$

### Bar and Cross Switching

$$P_1 = \sin^2(\frac{\phi}{2}), \quad P_2 = \cos^2(\frac{\phi}{2})$$

When $\phi = 0$, $\text{Output port } : \text{Output port 2} = 0 : 1$

When $\phi = \pi/2$, $\text{Output port 1} : \text{Output port 2} = 0.5 : 0.5$

When $\phi = \pi$, $\text{Output port 1} : \text{Output port 2} = 1 : 0$

## Switching Performance Metrics

Splitting ratio:
$$P_1/(P_1 + P_2)$$

Extinction ratio:
$$10\log_{10}(P_\text{max}/P_\text{min})$$

Insertion loss:
$$-10\log_{10}(P_\text{out, total}/P_\text{in, total})$$

In an ideal MZI, the output power at one port can be zero, meaning the extinction ratio can theoretically be infinite. Also, for an ideal, lossless MZI, the total insertion loss is $0\,\mathrm{dB}$, as the power is merely redistributed between the output ports.
We have to distinguish between the appearance of power at an unintended output port and the actual occurrence of optical loss.

## Nonideal MZI: Coupling Errors, Phase Errors, and Loss

Let's consider a non-ideal coupler:
$$\theta_1 = \frac{\pi}{4} + \delta\theta_1$$
$$\theta_2 = \frac{\pi}{4} + \delta\theta_2$$

The whole matrix is still:
$$U_\text{MZI}(\theta_1, \theta_2, \phi) = C(\theta_2)P(\phi)C(\theta_1) $$

Let's define the phase error,
$$\phi_\text{actual} = \phi_\text{command} + \delta\phi$$

Let's define the arm propagation matrix, which contains loss
$$D = \begin{bmatrix}
e^{-\alpha_1L_1/2}e^{i\phi_1} & 0 \\
0 & e^{-\alpha_2L_2/2}e^{i\phi_2}
\end{bmatrix}$$

$\alpha_j$ is the power decay constant, $\phi_j = -\beta_jL_j$

The total MZi is
$$U_\text{nonideal} = C_2DC_1$$

If loss occurs, it is generally not unitary.
$$U_\text{nonideal}^\dagger U_\text{nonideal} \ne I $$

## Numerical Study Case 1 — Ideal MZI: Transfer Matrix and Switching Verification

Can a transfer-matrix model reproduce the analytical switching response of an ideal lossless Mach–Zehnder interferometer?

$$C(\theta) = \begin{bmatrix}
\cos\theta & -i\sin\theta \\
-i\sin\theta & \cos\theta
\end{bmatrix}$$

$$P(\phi) = \begin{bmatrix}
1 & 0 \\
0 & e^{i\phi}
\end{bmatrix}$$

In general,
$$U_\text{MZI}(\theta_1, \theta_2, \phi) = C(\theta_2)P(\phi)C(\theta_1) $$

Reference condition:
$$\theta_1 = \theta_2 = \frac{\pi}{4}$$

So,
$$U_\text{MZI}(\theta_1, \theta_2, \phi) = C(\frac{\pi}{4})P(\phi)C(\frac{\pi}{4}) $$

for reference input,
$$a_\text{output} = \frac{1}{2}\begin{bmatrix}
1-e^{i\phi}\\
-i(1+e^{i\phi})
\end{bmatrix}
$$

then we can get:
$$P_1 = \sin^2(\frac{\phi}{2}), \quad P_2 = \cos^2(\frac{\phi}{2})$$

```python
def coupler_matrix(theta: float) -> np.ndarray:
    c = np.cos(theta)
    s = np.sin(theta)

    C = np.array([
        [c, -1j * s],
        [-1j * s, c]
    ], dtype=complex)

    return C
```

```python
def phase_matrix(phi: float) -> np.ndarray:
    P = np.array([
        [1, 0],
        [0, np.exp(1j * phi)]
    ], dtype=complex)

    return P
```

```python
def mzi_matrix(theta_1: float, theta_2: float, phi: float) -> np.ndarray:
    C_1 = coupler_matrix(theta_1)
    C_2 = coupler_matrix(theta_2)
    P = phase_matrix(phi)

    U = C_2 @ P @ C_1

    return U
```

```python
theta = np.pi / 4

C = coupler_matrix(theta)
print(C)

I = np.eye(2, dtype=complex)

unitarity_residual = np.linalg.norm(
    C.conj().T @ C - I,
    ord='fro'
)

print("Unitarity residual:", unitarity_residual)
```

```python
print("phi = 0")
print(phase_matrix(0))

print("phi = pi")
print(phase_matrix(np.pi))

phi = np.pi / 3

P = phase_matrix(phi)
I = np.eye(2, dtype=complex)

phase_unitarity_residual = np.linalg.norm(
    P.conj().T @ P - I,
    ord='fro'
)

print("Phase unitarity residual:", phase_unitarity_residual)
```

```python
theta_1 = np.pi / 4
theta_2 = np.pi / 4
phi = 0.0

U = mzi_matrix(theta_1, theta_2, phi)

a_in = np.array([1, 0], dtype=complex)
a_out = U @ a_in

print("MZI transfer matrix:")
print(U)

print("Output amplitudes:")
print(a_out)

print("Output powers:")
print(np.abs(a_out)**2)
```

```python
phi_values = np.linspace(0, 2 * np.pi, 201)

power_numerical = np.array([np.abs(mzi_matrix(theta_1, theta_2, phi) @ a_in)**2 for phi in phi_values])
power_analytical = np.column_stack([np.sin(phi_values / 2)**2, np.cos(phi_values / 2)**2])

max_error = np.max(np.abs(power_numerical - power_analytical))

print("Maximum absolute error:", max_error)
```

```python
I = np.eye(2, dtype=complex)

unitarity_errors = []

for phi in phi_values:
    U = mzi_matrix(theta_1, theta_2, phi)

    residual = np.linalg.norm( U.conj().T @ U - I, ord="fro")
    unitarity_errors.append(residual)

max_unitarity_error = np.max(unitarity_errors)

print("Maximum MZI unitarity residual:", max_unitarity_error)
```

```python
a_in_1 = np.array([1, 0], dtype=complex)

a_in_2 = np.array([1 + 1j, 2 - 1j], dtype=complex)
a_in_2 /= np.linalg.norm(a_in_2)

for name, a_in_test in [("Single-port", a_in_1),("Complex two-port", a_in_2)]:

    power_in = np.sum(np.abs(a_in_test)**2)
    errors = []

    for phi in phi_values:
        U = mzi_matrix(theta_1, theta_2, phi)
        a_out = U @ a_in_test

        power_out = np.sum(np.abs(a_out)**2)
        errors.append(abs(power_out - power_in))

    print(f"{name}: maximum power error = {np.max(errors):.6e}")
```

### Results Interpretation

**Physical Question**

Can the ideal MZI transfer-matrix model reproduce analytical switching behavior while preserving unitarity and total optical power?

**Numerical Results**

The MZI was modeled using two ideal 50:50 directional couplers and a lossless phase shifter:

$$U_{\mathrm{MZI}}=C(\frac{\pi}{4})P(\phi)C(\frac{\pi}{4}).$$

The relative phase was sampled at 201 points over $0\leq\phi\leq2\pi$

**Analytical agreement:**

The maximum absolute difference between the numerically calculated output powers and the analytical expressions was $8.88\times10^{-16}$.

**Unitarity:** 

The maximum Frobenius norm of $U^\dagger U-I$ was $9.46\times10^{-16}$

**Power conservation:** 

The maximum input-output power discrepancy was $6.66\times10^{-16}$ for both single-port and normalized complex two-port inputs.

**Switching states:** 

Cross, equal-splitting, and bar states were reproduced at $\phi=0,\frac{\pi}{2},\pi$, respectively, with a maximum absolute power error of $4.44\times10^{-16}$.

**Physical Interpretation**

The results confirm that the implemented transfer-matrix model reproduces the analytical switching response of an ideal lossless MZI to machine precision.

The relative phase controls the redistribution of optical power between the two output ports, while the total output power remains constant. The unitary transfer matrix also guarantees power conservation for arbitrary input vectors.

This verified model provides a baseline for investigating nonideal coupling angles and switching leakage in subsequent numerical studies.

**Limitations**

The agreement verifies the consistency and numerical implementation of the ideal transfer-matrix model, not the accuracy of a fabricated photonic device. Propagation loss, fabrication variations, phase-setting errors, and higher-order modal effects are excluded from this case.

## Numerical Study Case 2 — Geometry-Induced Coupling Errors and Switching Robustness

How do variations in directional-coupler waveguide gaps affect the switching performance of a Mach–Zehnder interferometer?

**Recall Notebook 10' conditions**

Under the 1D slab-waveguide reduced-order model, 
$$\lambda_0 = 2.0 \mathrm{\mu m}$$
$$\text{Core width} = 0.8 \mathrm{\mu m}$$
$$\text{Gap},\, g_0 = 0.5 \mathrm{\mu m}$$
$$n_\text{core} = 1.5, \quad n_\text{clad} = 1.0$$
$$\kappa(g_0) \approx 0.0880847 \mathrm{\mu m^{-1}}$$
$$\text{Coupling decay coefficient}, \,\gamma = 2.738058 \mathrm{\mu m^{-1}}$$

Let's **fix** the length to make $50:50$ splitting,
$$\kappa (g_0)L_\text{DC} = \frac{\pi}{4}$$

Therefore,
$$L_\text{DC} = \frac{\pi}{4\kappa (g_0)} \approx 8.92 \mathrm{\mu m}$$

this length $L_\text{DC}$ is independent from Gap.

```python
L_DC = np.pi / (4 * kappa0)

def kappa_from_gap(g):
    return kappa0 * np.exp(-gamma * (g - g0))

def theta_from_gap(g):
    return kappa_from_gap(g) * L_DC

print("Fixed coupler length:", L_DC)
print("Reference coupling angle:", theta_from_gap(g0))
```

```python
delta_gap_range = np.linspace(0, 0.05, 101)

common_mode = []
differential_mode = []

for dg in delta_gap_range:

    # Common-mode error
    theta1 = theta_from_gap(g0 + dg)
    theta2 = theta_from_gap(g0 + dg)

    # Differential-mode error
    theta3 = theta_from_gap(g0 + dg)
    theta4 = theta_from_gap(g0 - dg)

    common_leakages = []
    differential_leakages = []

    for phi, port in [(0, 0), (np.pi, 1)]:

        U_common = mzi_matrix(theta1, theta2, phi)
        U_diff = mzi_matrix(theta3, theta4, phi)

        power_common = np.abs(U_common @ a_in)**2
        power_diff = np.abs(U_diff @ a_in)**2

        common_leakages.append(power_common[port])
        differential_leakages.append(power_diff[port])

    common_mode.append(common_leakages)
    differential_mode.append(differential_leakages)

common_mode = np.asarray(common_mode)
differential_mode = np.asarray(differential_mode)
    
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(delta_gap_range, common_mode[:, 0],
        label="Common — Cross")

ax.plot(delta_gap_range, common_mode[:, 1],
        label="Common — Bar")

ax.plot(delta_gap_range, differential_mode[:, 0],
        label="Differential — Cross")

ax.plot(delta_gap_range, differential_mode[:, 1],
        label="Differential — Bar")

ax.set_xlabel(r"Gap variation $\delta g$ ($\mu$m)")
ax.set_ylabel("Switching leakage power")
ax.grid(alpha=0.3)
ax.legend()

plt.show()    
```

So,
$$P_\text{leak, cross} \approx [\theta''(g_0)]^2(\delta g)^4$$

### Limitations

1. 1D scalar waveguide approximation: `10_directional_couplers_and_coupled_mode_theory`의 FDM은 실제의 3D photonics device를 온전히 기술하지 않는다.


2. Exponential coupling approximation: 지수적 $\kappa (g)$ 관계는 검증한 구간에서 유효한 근사이며 모든 기하학적 상황에 대해 일반화할 수 없다. 


3. Model-level robustness: 이상적 lossless MZI와 대칭적 perturbation 가정 아래 도출된 결과이다.


4. Discretization Uncertainty: exponential approximation의 정확도와 회로 수준에서의 영향만을 확인하였으나, discretization uncertainty를 별도로 정량화하여 검증하지는 않았다.

## Numerical Study — Case 3: Phase Calibration and Switching Recovery in a Nonideal MZI

To what extent can phase tuning compensate for phase-setting errors and geometry-induced coupling imperfections in a Mach–Zehnder interferometer?

$$U_\text{MZI} = C(\theta_2)P(\phi_\text{actual})C(\theta_1)$$
$$\phi_\text{actual} = \phi_\text{command} + \delta\phi$$

**Parameter**

Input: $[1, 0]^T$

Propagation loss: Neglected

Coupler length: $L_\text{DC} = \frac{\pi}{4\kappa (g_0)} \approx 8.92 \mathrm{\mu m}$

**Metric**

$$J(\phi_\text{commamd}) = \left|\left[U_\text{MZI}(\phi_\text{command} + \delta\phi)\begin{pmatrix} 
1 \\
0
\end{pmatrix}\right]_1\right|^2$$

Calcultae:
$$\min_{\phi_\text{command}} J(\phi_\text{command})$$

---

### Experiment A — Phase OFfest Sensitivity and Calibration:
Ideal coupler with
$$\delta\phi = 2^\circ \quad \delta\phi = 2.3^\circ$$

$$\text{Coupling angles}\,\,: \theta_1 = \theta_2 = \frac{\pi}{4}$$


$$\text{Controllable parameter}: \phi_\text{command} = 0 ; \,\text{Cross state}$$


```python
phi_command = 0.0
del_phi_range = np.linspace(0, np.pi/36, 101)

theta = np.pi / 4
a_in = np.array([1, 0], dtype=complex)

P_leak_1 = []
P_out_2 = []

for delta in del_phi_range:

    phi_actual = phi_command + delta

    U = mzi_matrix(theta, theta, phi_actual)

    power = np.abs(U @ a_in)**2

    P_leak_1.append(power[0])
    P_out_2.append(power[1])

P_leak_1 = np.asarray(P_leak_1)
P_out_2 = np.asarray(P_out_2)


P_analytical = np.sin(del_phi_range / 2)**2

max_error = np.max(np.abs(P_leak_1 - P_analytical))

print("Maximum analytical error:", max_error)

plt.plot(
    del_phi_range,
    P_leak_1,
    label = r"$P_{leak}$")

plt.xlabel(r"Phase offset $\delta\phi$ (rad)")
plt.ylabel("Normalized leakage power")
plt.title("MZI Phase Error Sensitivity")

plt.grid()
plt.legend()
plt.tight_layout()

plt.show()
```

### Experiment C — Analytical Limits: general coupling angles and actual phase
When
$$0 < \theta_1, \, \theta_2 < \frac{\pi}{2}$$

and 
$$\phi = \phi_\text{actual}$$

$$U_\text{MZI}(\theta_1, \theta_2, \phi) = C(\theta_2)P(\phi)C(\theta_1) $$

Input vector is:
$$\mathbf{a}_\text{in} = \begin{bmatrix}
1 \\
0
\end{bmatrix}$$

We already know $U_\text{MZI}$
$$U_\text{MZI} = \begin{bmatrix}
\cos\theta_2 & -i\sin\theta_2 \\ 
-i\sin\theta_2 & \cos\theta_2
\end{bmatrix}\begin{bmatrix}
1 & 0 \\
0 & e^{i\phi}
\end{bmatrix}\begin{bmatrix}
\cos\theta_1 & -i\sin\theta_1 \\ 
-i\sin\theta_1 & \cos\theta_1
\end{bmatrix}
$$

We can calculate the $\mathbf{a}_\text{out}$,
$$\mathbf{a}_\text{out} = U_\text{MZI}\mathbf{a}_\text{in}$$
$$\mathbf{a}_\text{out} = \begin{bmatrix}
\cos\theta_2 & -i\sin\theta_2 \\ 
-i\sin\theta_2 & \cos\theta_2
\end{bmatrix}\begin{bmatrix}
1 & 0 \\
0 & e^{i\phi}
\end{bmatrix}\begin{bmatrix}
\cos\theta_1 & -i\sin\theta_1 \\ 
-i\sin\theta_1 & \cos\theta_1
\end{bmatrix}\begin{bmatrix}
1 \\
0
\end{bmatrix}$$

so,
$$\mathbf{a}_\text{out} = \begin{bmatrix}
\cos\theta_1\cos\theta_2 - \sin\theta_1\sin\theta_2e^{i\phi} \\
-i\sin\theta_2\cos\theta_1 - i\cos\theta_2\sin\theta_1e^{i\phi}
\end{bmatrix}$$

Therefore,
$$P_1(\phi) = \left|\cos\theta_1\cos\theta_2 - \sin\theta_1\sin\theta_2e^{i\phi}\right|^2 $$
$$P_2(\phi) = \left|i\sin\theta_2\cos\theta_1 + i\cos\theta_2\sin\theta_1e^{i\phi}\right|^2 $$

then we can get,
$$P_1(\phi) = \cos^2\theta_1\cos^2\theta_2 + \sin^2\theta_1\sin^2 \theta_2 -2\cos\theta_1\cos\theta_2\sin\theta_1\sin\theta_2\cos\phi$$
$$P_2(\phi) = \sin^2\theta_2\cos^2\theta_1 + \cos^2\theta_2\sin^2\theta_1 + 2\cos\theta_1\cos\theta_2\sin\theta_1\sin\theta_2\cos\phi$$

Furthermore,
$$P_1(\phi) + P_2(\phi) = 1$$

---

**Minimize the output power, $P_1$; $\cos\phi = 1$**

$$P_\text{cross,min} = \cos^2\theta_1\cos^2\theta_2 + \sin^2\theta_1\sin^2 \theta_2 -2\cos\theta_1\cos\theta_2\sin\theta_1\sin\theta_2$$

$$(\cos\theta_1\cos\theta_2 - \sin\theta_1\sin\theta_2)^2 = \cos^2\theta_1\cos^2\theta_2 + \sin^2\theta_1\sin^2 \theta_2 -2\cos\theta_1\cos\theta_2\sin\theta_1\sin\theta_2$$

$$\cos^2(\theta_1 + \theta_2) = (\cos\theta_1\cos\theta_2 - \sin\theta_1\sin\theta_2)^2 $$

$$P_\text{cross,min} = \cos^2(\theta_1 + \theta_2)$$

We can get 0 leakage when $\theta_1 + \theta_2 = \frac{\pi}{2}$

**Minimize the output power, $P_2$; $\cos\phi = -1$**

$$P_2(\phi) = \sin^2\theta_2\cos^2\theta_1 + \cos^2\theta_2\sin^2\theta_1 - 2\cos\theta_1\cos\theta_2\sin\theta_1\sin\theta_2$$

$$(\sin\theta_2\cos\theta_1 - \cos\theta_2\sin\theta_1)^2 = \sin^2\theta_2\cos^2\theta_1 + \cos^2\theta_2\sin^2\theta_1 - 2\cos\theta_1\cos\theta_2\sin\theta_1\sin\theta_2$$

$$\sin^2(\theta_1 - \theta_2) = (\sin\theta_2\cos\theta_1 - \cos\theta_2\sin\theta_1)^2$$

We can get 0 leakage when $\theta_1 = \theta_2$

When $$\theta_1 = \theta_2 = \frac{\pi}{4}$$ we can get 0 leakage between the two states.

## Results Interpretation

일정하게 발생하는 phase  offset을 보정할 수 있는지 확인하기 위해 Experiment A를 통해 실험하였다. $2^\circ$ offset에서는 $358^\circ$의 command phi로 leakage를 0에 가깝게 감소시킨 반면, $2.3^\circ$의 offset에서는 $1^\circ$의 sampling grid에 의해 이산적 residual leakage가 $6.85 \times 10^{-6}$ 남음을 확인하였다. 물리적 보상 가능성을 판단할 때, 수치해석적 finite grid의 한계를 구분해야 함을 확인하였다. 

Coupling angles의 변화가 optimal phase에서도 leakage를 발생시키는지 확인하기 위해 Experiment B를 진행하였다. 이로 인해 Cross-state minimum에서는 $0.02173\%$의 leakage가, Bar-state minimum에서는 $4.5817\%$의 leakage가 남는 것을 확인하였다. 위상 최적화 상태에서도 geometry-induced leakage가 남음을 확인하였다. 

Experiment B의 결과를 수식을 통해 Experiment C에서 확인한 결과, 완전한 단일 switching state를 구현할 수 있는 조건과 두 switching state를 온전히 구현할 수 있는 조건이 **서로 다르다**. 또한, Phase shifter는 두 경로의 상대 위상을 조절할 수 있지만, coupler geometry가 결정하는 진폭의 상대적인 크기까지 독립적으로 조절할 수는 없다. 따라서, 정밀한 위상 조정에도 간섭하는 두 진폭의 크기가 동일하지 않으면 완전한 상쇄 간섭을 얻을 수 없다.

Phase-only calibration can eliminate a constant phase-setting error in an ideal MZI, whereas geometry-induced coupling errors impose state-dependent leakage floors that generally cannot be removed by adjusting the relative phase alone.

### Limitations

- 실제 소자의 propagation loss, arm imbalance, phase noise를 시뮬레이션한 것은 아니다.

- Geometry error는 `Notebook 10`의 exponential coupling approximation을 기반으로 한다.


- Analytical prediction과 numerical result의 일치는 모델 내부의 검증일 뿐, 독립적인 device-level validation이 아니다.
