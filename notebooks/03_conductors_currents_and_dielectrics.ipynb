# Conductors, Currents, and Dielectrics
How does matter respond to an applied electric field?

## Conductor



Conductor contains charges that can move through the material. When an external electric field is applied, these free charges move until the electric field produced by the redistributed charges cancels the electric field inside the conductor.

$$\mathbf{E} = 0, \quad \rho _{f, \text{bulk}} = 0$$

In electrostatic equilibrium, the electric field and the excess free volume-charge density vanish inside the bulk of a conductor.

This result applies inside the conducting material under electrostatic equilibrium.
We alread know $$\mathbf{E} = -\nabla V$$
So, the whole conductor is equipotential. It doesn't mean always $V=0$, It means the electric potential is constant over the space.

### Boundary condition at the conductor surface

$$\mathbf{E}_{\text{in}} = 0$$

### Electrostatic Equilibrium versus Steady Current

|Electrostatic equilibrium|Steady conduction|
|:-----------------:|:------------------:|
|Macroscopic redistribution of free charges is complete|charges are move continuously|
|$\mathbf{E}_{\text{in}} = 0$|$\mathbf{E}_{\text{in}} \neq 0$|
|no current|It allow steady current|
|The potential is constant|It allows a decrease in electric potential followed by a conductor|

### current

$\sigma \,$: Degree of electric flow in $[\mathrm{\ohm^-1 /m}]$

$\text{Current}$: $I$ Electric charge passing through a surface per unit time in $[\mathrm{c/sec}], [A]$
$$I = \frac{dQ}{dt}$$ $I$ doesn't care about the volume.

### current density

When the electric charge is in free space,





$$I = \frac{d Q}{d t}, \quad d Q = \rho _v \times d A \, d y$$
Therefore,
$$I = \frac{\rho _v \times d A d y}{d t} = \rho _v \, d A \frac{d y}{d t} = \rho _v \, d A \, u_y$$
Furthermore,
$$\frac{d I}{d A} = \rho _v u_y = J_y$$

$$\mathbf{J} = \rho _v \mathbf{u} \,\, [\mathrm{A/m^2}], \quad \rho _v = nQ_e$$

in general,
$$I = \int_s \mathbf{J} \cdot d \mathbf{A} $$

The drift velocity of electrons is opposite to the conventional current direction because the electron charge is negative.

### local or differential form of Ohm’s law

In a metal, the positive ion lattice remains approximately fixed, while conduction electrons acquire a small average drift velocity under an applied electric field $\vec{E}$. We can model the statistical collision, 
$$\mathbf{J} = \sigma E \quad \text{or} \quad \mathbf{E} = \rho _e \mathbf{J}$$ 



  
Between the two ends of this conductor is an electric potential difference of about $V$
It makes a steady current. When the magnitude of the electric field is homogeneous, we can write $$E = \frac{V}{L}$$, and also if this conductor has a homogeneous cross area, we can write $$J = \frac{I}{A} = \sigma E = \sigma \frac{V}{L}$$
Furthermore, 
$$\frac{I}{A} = \sigma \frac{V}{L} $$ 
$$ V = \rho_e \frac{L}{A}I \quad \rho _e = \frac{1}{\sigma}$$

in general,

$$R = \frac{V}{I} = \frac{\int_l \vec{\mathbf{E}} \cdot d\vec{l}}{\int_s \mathbf{j} \cdot d\mathbf{A}} = \frac{\int_l \vec{\mathbf{E}} \cdot d\vec{l}}{\int_ s \sigma \vec{\mathbf{E}} \cdot d\vec{A}}$$

$\sigma$: electrical conductivity

$\rho _e$: electrical resistivity

$$dU = -\mathbf{F} \cdot d\mathbf{l} = - q\mathbf{E} \cdot d\mathbf{l}$$
$$P = \frac{dw}{dt} = -\mathbf{F} \cdot \frac{d\mathbf{l}}{dt} = -\mathbf{F} \cdot \mathbf{v}$$
$$dP = (dQ\mathbf{E}) \cdot \mathbf{v} = (\rho _v dx dy dz \mathbf{E}) \cdot \mathbf{v} = \mathbf{E} \cdot \rho _v \mathbf{v} dx dy dz = \mathbf{E} \cdot \mathbf{J} dx dy dz$$
$$P = \iiint \mathbf{E} \cdot \mathbf{J} \, dv$$

If the conductor is homogeneous, 
$$P = \iiint \mathbf{E} \cdot \mathbf{J} \, dv = \int E dl \iint J ds = VI$$

## Dielectric

|Conductor|Dielectric|
|:-----------------------------:|:-----------------------------:|
|Free charges can move from a macroscopic perspective.|Electric charge is mainly bound to atoms or molecules.|
|$E_\text{in} = 0$|In general, $E \neq 0$|
|Free charges on the surface are redistributed.|Electric dipoles are induced or permanent dipoles partially align with the external electric field|
|free-charge response|bound-charge response|

### dipole moment and polarization



When charges $+q, -q$ separate by a displacement vector of $\mathbf{d}$,

$$\mathbf{p} = q \, \mathbf{d}$$

In this case, the direction of $\mathbf{d}$ must be from the negative charge to the positive charge.
Polarization has two mechanisms.

1. **Induced polarization**: The external field changes the relative positions of the electron cloud and the atomic nucleus.




2. **Orientational polarization**: Molecules that already possess permanent dipoles align in the direction of the external field.



$$\mathbf{P} = \lim_{\Delta v \to 0} \frac{\sum_{k=1}^{N} \mathbf{p}_k}{\Delta V}$$

### bound charge
Surface bound charge: $$\sigma _b = \mathbf{P} \cdot \hat{\mathbf{n}}$$

volume bound charge: $$\rho _b = -\nabla \cdot \mathbf{P}$$

### Electric displacement field
$$\mathbf{D} = \epsilon _0\mathbf{E} + \mathbf{P}$$

Gauss's law for free charges is as follows,
$$\nabla \cdot \mathbf{D} = \rho _f $$
In integral form,
$$\oint_{s} \mathbf{D} \cdot d{\mathbf{A}} = Q_{f,enc}$$

$\mathbf{E}$: The actual electric field created by all charges

$\mathbf{D}$: An auxiliary field introduced to organize material reactions based on free charge.

$\mathbf{P}$: Internal dipole moment density of a material

### Linear dielectric

These relations assume a linear, isotropic, and homogeneous dielectric.
$$\mathbf{P} = \epsilon _0 \chi_e \mathbf{E}$$

Therefore, 
$$\mathbf{D} = \epsilon_0(1+\chi_e)\mathbf{E} = \epsilon \mathbf{E}$$
$$\epsilon = \epsilon _0 \epsilon _r$$
$$\epsilon _r = 1 + \chi_e$$

### Physical interpretation of dielectric screening



When an electric field generated by an external capacitor is applied to a dielectric, atoms and molecules within the dielectric become polarized, and bound charges appear at the interface. For a fixed free-charge distribution, the field produced by bound charges partially opposes the original field and reduces the net electric field inside the dielectric.

### Concept Check 1

A neutral conductor is placed in a uniform external electric field.

1. Does the conductor remain locally neutral everywhere?
2. Is the electric field inside the conducting material zero immediately?
3. What produces the field that cancels the external field?
4. Is the potential of the conductor necessarily zero?

#### Answer
1. 아니다. 전체 순전하는 0이지만, 표면에서 국소적인 전하 불균형이 발생한다.
   
    No. Although the conductor remains neutral as a whole, induced positive and negative surface-charge regions appear on opposite sides. The bulk interior remains approximately neutral.

2. 도체가 perfect conductor일 경우 즉각적으로 $\mathbf{E} = 0, \rho _v = 0$이 되나, 이외의 경우에는 charge relaxation time 동안 전하가 재배치된다.

    Not necessarily immediately. In a real conductor, charges require a finite relaxation time to redistribute. The instantaneous response of a perfect conductor is an idealization.

3. 재배치된 표면 전하들이 외부장을 상쇄시킨다.

    The redistributed surface charges produce an induced electric field that cancels the external field inside the conductor.
 
4. 정전기적 평형 상태에 있는 도체의 임의의 두 점 사이의 전위차가 0이어야 할 뿐, 전위 자체의 값이 0이어야 할 필요는 없다.

    No. The potential of the conductor need not be zero. It must be spatially constant, so the potential difference between any two points of the same conductor is zero. Its absolute value depends on the chosen reference and on whether the conductor is grounded.


### Concept Check 2

1. Why must the tangential component of the electric field vanish at the
surface of a conductor in electrostatic equilibrium?

#### Answer

1. 장의 접선 방향, 즉 표면에 접하는 힘이 사라져야 하는 이유는 장과 평행한 힘은 자유전하를 움직여 정전기적 평형 상태를 유지하지 못하게 되기 때문이다.

    If a tangential electric-field component existed, it would exert a tangential force on the free surface charges. The charges would continue to move, contradicting electrostatic equilibrium.


### Concept Check 3

A metal wire carries a steady current.

1. Is the electric field inside the wire necessarily zero?
2. Are the electrons moving at the same speed as the electrical signal?
3. How are the electron drift direction and conventional current direction related?
4. What physical quantity does current density describe?

#### Answer

1. 아니다. 전선 내부에 전자를 밀어주는 일정한 전기장이 존재한다.


   No. A nonzero electric field is generally required inside a resistive wire to maintain a steady conduction current.
   
2. 전자 자체의 평균 drift speed는 느리고, 전기장의 변화 또는 신호 전달은 빠르다.

    No. The electron drift velocity is typically very small, whereas the electromagnetic signal propagates through the circuit at a substantial fraction of the speed of light in the surrounding medium.

3. 전류의 방향의 반대 방향으로 전자가 이동한다.

    In an ordinary metal, the electron drift velocity is opposite to the conventional-current direction because electrons carry negative charge.

4. metal wire에 존재하는 자유전하의 drift motion에 의해, current density는 전도 전류를 나타내며, 이를 식으로 쓰면 $$dI = \mathbf{J} \cdot d\mathbf{A}, \quad I = \int_s \mathbf{J} \cdot d \mathbf{A}$$ 이다.

   Current density describes the amount and direction of charge flow per unit area perpendicular to the flow.

## Case 1: Current and Joule Heating in a Uniform Conducting Wire

### Assumptions

- steady conduction
- uniform and isotropic conductor
- constant conductivity
- uniform cross-sectional area
- negligible temperature dependence
- negligible contact resistance
- no charge accumulation in the bulk

### Physical system
$$L = 1.0 \mathrm{m}$$
$$a = 1.0 \mathrm{mm}$$
$$\sigma = 5.8 \times 10^7 \mathrm{S/m}$$
$$V_0 = 1.0 \, \mathrm{V}$$




```python
L = 1.0 # [m]
a = 1.0 * 10**(-4) # [mm]
sigma = 5.8 * 10**7 # [S/m]
voltage_0 = 1.0 # [V]

response = uniform_wire_response(voltage_0, L, a, sigma)

inner_electric_field = response["inner_electric_field"]
current_density = response["current_density"]
area = response["area"]
current = response["current"]
resistance = response["resistance"]
power = response["power"]
volumetric_power_density = response["volumetric_power_density"]
```

```python
r_I = abs(current - (voltage_0 / resistance)) / abs((voltage_0 / resistance))
r_P = abs(power - (current**2)*resistance) / abs((current**2)*resistance)
r_P_volume = abs(power - (current_density*inner_electric_field*area*L)) / abs((current_density*inner_electric_field*area*L))
```

```python
x_1 = np.linspace(0.5, 2.0, 301)

R = np.zeros_like(x_1)
I = np.zeros_like(x_1)
P = np.zeros_like(x_1)

for i, a_value in enumerate(x_1):
    response_a = uniform_wire_response(voltage_0, L, a_value, sigma)
    
    try:
        R[i] = response_a["resistance"]
        I[i] = response_a["current"]
        P[i] = response_a["power"]

    except ValueError:
        R[i] = np.nan
        I[i] = np.nan
        P[i] = np.nan



#R
plt.figure(figsize=(6, 4))

plt.plot(x_1, R, label="Resistance", color = "green")

plt.xlim(0.5, 2.0)

plt.xlabel("Wire radius a(mm)")
plt.ylabel(r"Resistance $R$ ($\Omega$)")
plt.title("Resistance as a Function of Wire Radius")
plt.grid(True)
plt.legend()
plt.show()


#I
plt.figure(figsize=(6, 4))

plt.plot(x_1, I, label="Current")

plt.xlim(0.5, 2.0)

plt.xlabel("Wire radius a(mm)")
plt.ylabel(r"Current $I$ ($V$)")
plt.title("Current as a Function of Wire Radius")
plt.grid(True)
plt.legend()
plt.show()   


#P
plt.figure(figsize=(6, 4))

plt.plot(x_1, P, label="Power", color = "skyblue")

plt.xlim(0.5, 2.0)

plt.xlabel("Wire radius a(mm)")
plt.ylabel(r"Power $P$ ($W$)")
plt.title("Power as a Function of Wire Radius")
plt.grid(True)
plt.legend()
plt.show()    
```
