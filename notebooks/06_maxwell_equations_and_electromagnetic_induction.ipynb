# Maxwell Equations and Electromagnetic Induction

How do Maxwell's equations unify electrostatics, magnetostatics, and electromagnetic induction into a complete description of classical electromagnetism?

## Why Maxwell's Equations?

- Gauss's law: electric flux is related to enclosed charge
- magnetostatics: currents generate magnetic fields
- Gauss's law for magnetism: No magnetic monopoles have been observed
- Faraday's law: A changing magnetic field generates electric fields
- Ampère-Maxwell Law: current and changing electric fields generate magnetic fields

Maxwell's equations combine these previously separate observations into a self-consistent field theory of electromagnetism.


## The Four Maxwell Equations
### Gauss's Law for Electricity

Differential form:
$$\nabla \cdot \mathbf{D} = \rho _v$$

Integral form:
$$\oint_S \mathbf{D} \cdot d\mathbf{A} = Q_enc$$

**Physical Interpretation**

Electric charge acts as a source or sink of the electric flux density $\mathbf{D}$

In a vacuum, we can also write the differential form,
$$\nabla \cdot \mathbf{E} = \frac{\rho}{\epsilon _0}$$ and $\nabla \cdot \mathbf{E}$ express local source density.

### Gauss's Law for Magnetism

Differential form:
$$\nabla \cdot \mathbf{B} = 0$$

Integral form:
$$\oint_{S} \mathbf{B} \cdot d\mathbf{A} = 0$$

**Physical Interpretation**

The magnetic field has no known isolated sources or sinks. Magnetic field lines therefore form closed loops.

### Faraday's Law

Differential form:
$$\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}$$

Integral form:
$$\oint_C \mathbf{E} \cdot d\mathbf{l} = - \frac{d}{dt}\int_S \mathbf{B} \cdot d\mathbf{A}$$


|Electrostatic field|Time-varying electromagnetic field|
|:----------------------:|:----------------------:|
|$\frac{\partial \mathbf{B}}{\partial t} = 0 \to \nabla \times \mathbf{E} = 0$|$\frac{\partial \mathbf{B}}{\partial t} \neq 0 \to \nabla \times \mathbf{E} \neq 0$|

**Physical Interpretation**

A time-varying magnetic field generates a circulating electric field.

### Ampère-Maxwell Law

Differential form:

$$\nabla \times \mathbf{H} = \mathbf{J}_f + \frac{\partial \mathbf{D}}{\partial t}$$

Integral form:
$$\oint_C \mathbf{H} \cdot d\mathbf{l} = \int_S \mathbf{J}_f \cdot d\mathbf{A} + \frac{\partial}{\partial t}\int_S \mathbf{D} \cdot d\mathbf{A} $$

**Physical Interpretation**



## Static Limits of Maxwell's Equations

In electrostatics, 
$$\frac{\partial \mathbf{B}}{\partial t} = 0$$
so,
$$\nabla \times \mathbf{E} = 0$$
Therefore,
$$\nabla \cdot \mathbf{D} = \rho _v, \quad \nabla \times \mathbf{E} = 0$$
In magnetostatics, since there is no change in time,
$$\frac{\partial \mathbf{D}}{\partial t} = 0$$
so, 
$$\nabla \times \mathbf{H} = \mathbf{J} _f, \quad \nabla \cdot \mathbf{B} = 0$$

Electrostatics and magnetostatics can therefore be interpreted as static limits of the full Maxwell equations.

## Constitutive Relations

$$\mathbf{D} = \epsilon \mathbf{E}$$
$$\mathbf{B} = \mu \mathbf{H}$$
$$\mathbf{J} = \sigma \mathbf{E}$$

Maxwell's equations describe the electromagnetic fields, while constitutive relations specify how a material responds to those fields.

## Current Continuity Equation

$$\nabla \cdot \mathbf{J} = -\frac{\partial \rho _v}{\partial t}$$

From the original Ampère's law,
$$\nabla \times \mathbf{H} = \mathbf{J}$$

Taking the divergence of both sides gives
$$\nabla \cdot (\nabla \times \mathbf{H}) = \nabla \cdot \mathbf{J}$$ 

According to a vector calculus identity, the divergence of a curl must be zero. 

Therefore, 
$$\nabla \cdot (\nabla \times \mathbf{H}) = 0 = \nabla \cdot \mathbf{J}$$

However, from the continuity equation, 
$$\nabla \cdot \mathbf{J} = -\frac{\partial \rho _v}{\partial t}$$

Therefore, in a time-dependent case where the charge density changes with time,
$$\nabla \cdot (\nabla \times \mathbf{H}) = 0 \neq \nabla \cdot \mathbf{J} = -\frac{\partial \rho _v}{\partial t}$$ 
which produces a contradiction. 

The displacement-current term was introduced to resolve this inconsistency. 

After introducing the displacement-current term,
$$\nabla \times \mathbf{H} = \mathbf{J}_f + \mathbf{J}_d$$taking the divergence gives
$$\nabla \cdot (\nabla \times \mathbf{H}) = \nabla \cdot \mathbf{J}_f + \nabla \cdot \mathbf{J}_d = 0$$
Rearranging this expression gives
$$-\frac{\partial \rho _v}{\partial t} + \nabla \cdot \mathbf{J}_d = 0$$
and therefore,
$$\nabla \cdot \mathbf{J}_d = \frac{\partial \rho _v}{\partial t} = \frac{\partial}{\partial t}\nabla \cdot \mathbf{D}$$

Maxwell's correction is introduced in the form
$$\mathbf{J}_d = \frac{\partial \mathbf{D}}{\partial t} $$
which makes the Ampère-Maxwell law consistent with charge conservation.

Therefore, Maxwell's equations and the constitutive relations together determine the electromagnetic response of a material system.

## From Maxwell's Equations to Electromagnetic Waves

In source-free homogeneous medium, 
$$\rho _v  = 0, \quad \mathbf{J} _f = 0$$

In linear medium,
$$\mathbf{D} = \epsilon \mathbf{E}, \quad \mathbf{B} = \mu \mathbf{H}$$

Let's derive the electric field differential equation.
$$\nabla \times \mathbf{E} = - \frac{\partial \mathbf{B}}{\partial t}$$

Taking the curl of both sides gives
$$\nabla \times (\nabla \times \mathbf{E}) = \nabla \times \left(-\frac{\partial \mathbf{B}}{\partial t}\right)$$
Because of, 
$$\nabla \times (\nabla \times \mathbf{E}) = \nabla(\nabla \cdot \mathbf{E}) - \nabla ^2\mathbf{E}$$
and in sourece-free homogeneous medium, 
$$\nabla \cdot \mathbf{D} = 0$$
and the $\epsilon$ is constant,
$$\nabla \cdot \mathbf{E} = 0$$
so,
$$-\nabla ^2\mathbf{E} = \frac{\partial}{\partial t}\left(\mu \epsilon  \frac{\partial \mathbf{E}}{\partial t}\right) = -\mu  \epsilon  \frac{\partial ^2 \mathbf{E}}{\partial t^2}$$

Therefore,
$$\nabla ^2\mathbf{E} - \mu\epsilon\frac{\partial ^2 \mathbf{E}}{\partial t^2} = 0$$


Similarly, In the same homogeneous source-free medium, we can get 

$$\nabla \times \mathbf{H} = \frac{\partial \mathbf{D}}{\partial t}$$

$$\nabla ^2\mathbf{B} - \mu\epsilon\frac{\partial ^2 \mathbf{B}}{\partial t^2} = 0$$
