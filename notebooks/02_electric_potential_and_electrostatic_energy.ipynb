# Electric Potential and Electrostatic Energy

## Electric potential and Potential Difference

When the test charge $q_0$ is in the electric field $\vec{\mathbf{E}}\,$, the test charge is influenced by the electric force, $q_0\vec{\mathbf{E}}\,$. This test charge is written by Coulomb's law, $q_0\vec{\mathbf{E}}\,$ which is a conservative force. In the charge-electric field system, the work that the electric field did on the charge is $$W=\vec{\mathbf{F}}_e \cdot d\vec{s} = q_0\,\vec{\mathbf{E}}\cdot d\vec{s}$$

As electric field work, electric potential energy charge as much $dU = -q_0\vec{\mathbf{E}}\cdot d\vec{s}$.

Therefore,
$$\Delta U = -q_0\int_{1}^{2}\vec{\mathbf{E}}\cdot d\vec{s}$$

$q_0\vec{\mathbf{E}}\,$ is hhe electrostatic force is conservative, so $\Delta U$ irrelevant to the route.

When the test charge is located in the electric field, it has a potential $U$ relative to the system at infinity, where $U=0$. When we divide $U$, we obtain a physical quantity that **depends only on the source charge distribution**. We call this physical quantity the electric potential, $V$
$$V=\frac{U}{q_0} \, [\mathrm{J/C}]$$

So, we can calculate the difference between two locations A, B in the electric field.
$$\Delta V = \frac{\Delta U}{q_0} = -\int_{A}^{B}\vec{\mathbf{E}}\cdot d\vec{s}$$

When the test charge moves from A to B with constant kinetic energy, then the external force did work to make the point charge's electric potential energy.
$$W_\text{ext}=\Delta U = q\Delta V$$

## Electric Potential and Potenctial Energy Due to Point Charges

For the electric potential under the electric field created by the point charge, let's use the general equation for potential difference.

$$\Delta V = -\int_{A}^{B}\vec{\mathbf{E}}\cdot d\vec{s}$$

And we can write the electric field with Coulomb's law
$$\vec{\mathbf{E}}\cdot d\vec{{s}} = k_e\frac{q}{r^2}\hat{\mathbf{r}}\cdot d{\vec{{s}}}$$
$\left| \hat{\mathbf{r}} \right| = 1$, 
$$\hat{\mathbf{r}} \cdot d\vec{s} = ds \cos\theta $$ $$ds \cos\theta = dr $$

If we substitute this into the above equation,

$$\Delta V = -k_e q\int_{r_A}^{r_B}\frac{dr}{r^2}$$
$$\Delta V = -k_e q[\frac{1}{r_A}-\frac{1}{r_B}]$$

Let's set the $V_A$ to reference point, $r_A = r_\infty = \infty$
$$V - V_{A} = -k_e q[\frac{1}{r_\infty}-\frac{1}{r}] $$ 
Therefore, we can get a general equation about electric potential $V$,
$$V = k_e q\frac{1}{r} $$

Furthermore,

Because of Principle of superposition,

$$V = k_e \sum_{i}\frac{q_i}{r_i}$$

## Obtaining the Value of the Electric Field from the Electric Potential
From $$\Delta V = -q\int_{A}^{B}\vec{\mathbf{E}}\cdot d\vec{s}$$

we can get $$dV = -\vec{\mathbf{E}}\cdot d\vec{{s}}$$
Let's describe $dV$ with vector form
$$ dV = [\frac{\partial V}{\partial x}dx +\frac{\partial V}{\partial y}dy + \frac{\partial V}{\partial z}dz] $$
Let's describe $\vec{\mathbf{E}}$ with vector form
$$ \vec{\mathbf{E}}\cdot d\vec{{s}} = \vec{\mathbf{E}}_x dx + \vec{\mathbf{E}}_y dy + \vec{\mathbf{E}}_z dz$$
If we organize this,
$$\vec{\mathbf{E}}_x = - \frac{\partial V}{\partial x} \quad \vec{\mathbf{E}}_y = - \frac{\partial V}{\partial y}\quad \vec{\mathbf{E}}_z = - \frac{\partial V}{\partial z}$$

$$\vec{\mathbf{E}} = -(\frac{\partial V}{\partial x}\hat{i}+\frac{\partial V}{\partial y}\hat{j}+\frac{\partial V}{\partial z}\hat{k})$$

$$\mathbf{E} = -\nabla V$$

## Electric Potential Due to Continuous Charge Distributions

We can calculate the electric potential of a continuous charge distribution. If we know the distribution of continuous charge, 

We can write the equation with reference point $P$
$$dV = k_e \frac{dq}{r}$$
$$V = k_e \int\frac{dq}{r}$$

and also we can use Gauss's law.

## Physical interpretations with answering the questions.

Q1. Why is the electric potential zero at the origin?

A1. 전위는 $$V = k_e q\frac{1}{r} $$로 정의되고, 주어진 두 전하는 전하량의 크기가 같으나 부호가 다르고, 원점으로 부터 떨어진 거리가 같기에, 원점에 대해 각 전하가 생성하는 전위의 크기가 같고 부호가 다르다. 따라서 원점에서의 알짜 전위는 0이 된다.

---

Q2. Why is the electric field nonzero at the origin?

A2. 두 전하량의 부호가 다르므로, 두 전하 사이에서 두 전하가 생성하는 전기장의 방향은 서로 동일하다. 이는 `01_electrostatic_force_and_electric_field`의 `Case 3`의 그래프와 `Physical interpretations and answering questions`에서 자세히 확인할 수 있다.

---

Q3. Why is the entire $y$-axis an equipotential line with (V=0)?

A3. 주어진 두 전하는 전하량이 같고 부호가 다르며, 원점으로부터 떨어진 거리가 동일하다. 따라서 $x=0$ 선상에 존재하는 모든 점들에서 각 전하와의 거리를 구할 때, 이 모든 거리는 각 전하에서 동일하다. Q1과 마찬가지로 전위는 $$V = k_e q\frac{1}{r} $$로 정의되고, 각 전하가 생성하는 전위의 크기가 같고 부호가 다르다. 따라서 $x=0$ 선상에서의 알짜 전위는 0이 된다.

---

Q4. In which direction does the electric field point relative to increasing and decreasing potential?

A4. 전기장은 전위가 감소하는 방향을 향한다. $$\mathbf{E} = -\nabla V$$ 이기에, 전기장은 전위가 가장 가파르게 낮아지는 방향을 향한다.

---

Q5. Why does zero electric potential not imply zero electric field?

A5. 전위는 스칼라 이지만 전기장은 벡터이기 때문이다.

## Case 3. Reconstructing the Electric Field from the Potential

A single positive point charge is located at the origin:

$$q=1.0\times10^{-9}\ \mathrm{C}$$

Construct a two-dimensional computational domain over

$$-0.20\le x\le0.20\ \mathrm{m}, \qquad -0.20\le y\le0.20\ \mathrm{m}$$

Physical Question

Does the numerical relation

$$\mathbf E=-\nabla V$$

reproduce the electric field obtained directly from Coulomb’s law?

### Tasks
1. Construct a uniform two-dimensional grid.


2. Calculate the electric potential at every grid point 


3. Exclude the charge position because the potential and electric field are undefined there.

4. Use 'numpy.gradient' to reconstruct the electric-field components:

$$-\frac{\partial V}{\partial y}$$

5. Calculate the electric field directly from Coulomb’s law:

$$k_eq\frac{\mathbf r}{|\mathbf r|^3}$$

6. Calculate the absolute error:

$$\Delta E = \left| \mathbf E_{\mathrm{gradient}} - \mathbf E_{\mathrm{direct}} \right|$$

7. Calculate the relative error where the direct electric-field magnitude is sufficiently large:

$$\epsilon_{\mathrm{rel}} =\frac{\left| \mathbf E_{\mathrm{gradient}} - \mathbf{E}_{\mathrm{direct}} \right|}{\left| \mathbf E_{\mathrm{direct}} \right|
}$$

8. Exclude the following regions from the error analysis:

* the charge position and nearby grid points,
* the outer boundary of the computational domain,
* points where the direct electric-field magnitude is extremely small.

9. Plot the reconstructed electric field or the error distribution.
10. Repeat the calculation using at least two different grid resolutions and compare the numerical errors.

### Verification

Check whether the reconstructed field

* points radially outward from the positive charge,
* agrees with the direct field away from the charge and boundaries,
* becomes more accurate as the grid resolution increases.

```python
q_3 = 1.0e-9
charge_position_3 = np.array([0.0, 0.0])

x_3 = np.linspace(-0.2, 0.2, 101)
y_3 = np.linspace(-0.2, 0.2, 101)

X_3, Y_3 = np.meshgrid(x_3, y_3)

dx_3 = x_3[1] - x_3[0]
dy_3 = y_3[1] - y_3[0]
```

```python
V_3 = np.zeros_like(X_3, dtype=float)

for i in range(X_3.shape[0]):
    for j in range(X_3.shape[1]):
        observation_point = np.array([
            X_3[i, j],
            Y_3[i, j]
        ])

        try:
            V_3[i, j] = electric_potential_point_charge(
                observation_point,
                charge_position_3,
                q_3
            )

        except ValueError:
            V_3[i, j] = np.nan
```

```python
V_for_gradient_3 = V_3.copy()

nan_mask_3 = np.isnan(V_for_gradient_3)

nearest_distance_3 = min(dx_3, dy_3)

replacement_potential_3 = K_e * q_3 / nearest_distance_3

V_for_gradient_3[nan_mask_3] = replacement_potential_3

dV_dy_3, dV_dx_3 = np.gradient(
    V_for_gradient_3,
    dy_3,
    dx_3
)

E_x_gradient_3 = -dV_dx_3
E_y_gradient_3 = -dV_dy_3
```

```python
E_x_direct_3 = np.zeros_like(X_3, dtype=float)
E_y_direct_3 = np.zeros_like(Y_3, dtype=float)

for i in range(X_3.shape[0]):
    for j in range(X_3.shape[1]):
        observation_point = np.array([
            X_3[i, j],
            Y_3[i, j]
        ])

        try:
            electric_field_3 = electric_field_point_charge(
                observation_point,
                charge_position_3,
                q_3
            )

            E_x_direct_3[i, j] = electric_field_3[0]
            E_y_direct_3[i, j] = electric_field_3[1]

        except ValueError:
            E_x_direct_3[i, j] = np.nan
            E_y_direct_3[i, j] = np.nan
```

```python
right_point_index_3 = np.argmin(
    (X_3 - 0.1)**2 + Y_3**2)

right_i_3, right_j_3 = np.unravel_index(
    right_point_index_3,
    X_3.shape)
```

```python
absolute_error_3 = np.sqrt((E_x_gradient_3 - E_x_direct_3)**2 + (E_y_gradient_3 - E_y_direct_3)**2)

E_direct_magnitude_3 = np.sqrt(E_x_direct_3**2 + E_y_direct_3**2)
```

```python
relative_error_3 = np.full_like(absolute_error_3, np.nan, dtype=float)

R_3 = np.sqrt(X_3**2 + Y_3**2)

charge_exclusion_radius_3 = 3.0 * max(dx_3, dy_3)

valid_mask_3 = R_3 > charge_exclusion_radius_3

boundary_mask_3 = np.ones_like(X_3, dtype=bool)

boundary_mask_3[:2, :] = False
boundary_mask_3[-2:, :] = False
boundary_mask_3[:, :2] = False
boundary_mask_3[:, -2:] = False

valid_mask_3 &= boundary_mask_3

valid_mask_3 &= np.isfinite(absolute_error_3)
valid_mask_3 &= np.isfinite(E_direct_magnitude_3)

field_threshold_3 = 1.0e-12

valid_mask_3 &= E_direct_magnitude_3 > field_threshold_3

relative_error_3[valid_mask_3] = (absolute_error_3[valid_mask_3] / E_direct_magnitude_3[valid_mask_3])
```

```python
mean_relative_error_3 = np.nanmean(relative_error_3[valid_mask_3])

median_relative_error_3 = np.nanmedian( relative_error_3[valid_mask_3])

maximum_relative_error_3 = np.nanmax(relative_error_3[valid_mask_3])
```

```python
def calculate_reconstruction_error(grid_points: int) -> tuple[float, float]:
    x = np.linspace(-0.2, 0.2, grid_points)
    y = np.linspace(-0.2, 0.2, grid_points)

    X, Y = np.meshgrid(x, y)

    dx = x[1] - x[0]
    dy = y[1] - y[0]

    R = np.sqrt(X**2 + Y**2)

    V = np.full_like(R, np.nan, dtype=float)

    finite_mask = R > 0.0
    V[finite_mask] = K_e * q_3 / R[finite_mask]

    V_for_gradient = V.copy()
    V_for_gradient[~finite_mask] = (
        K_e * q_3 / min(dx, dy)
    )

    dV_dy, dV_dx = np.gradient(
        V_for_gradient,
        dy,
        dx
    )

    E_x_gradient = -dV_dx
    E_y_gradient = -dV_dy

    E_x_direct = np.full_like(X, np.nan, dtype=float)
    E_y_direct = np.full_like(Y, np.nan, dtype=float)

    E_x_direct[finite_mask] = (
        K_e * q_3
        * X[finite_mask]
        / R[finite_mask]**3
    )

    E_y_direct[finite_mask] = (
        K_e * q_3
        * Y[finite_mask]
        / R[finite_mask]**3
    )

    absolute_error = np.sqrt(
        (E_x_gradient - E_x_direct)**2
        +
        (E_y_gradient - E_y_direct)**2
    )

    direct_magnitude = np.sqrt(
        E_x_direct**2
        +
        E_y_direct**2
    )

    valid_mask = R > 0.02

    valid_mask[:2, :] = False
    valid_mask[-2:, :] = False
    valid_mask[:, :2] = False
    valid_mask[:, -2:] = False

    valid_mask &= np.isfinite(absolute_error)
    valid_mask &= direct_magnitude > 0.0

    relative_error = (
        absolute_error[valid_mask]
        /
        direct_magnitude[valid_mask]
    )

    return (
        np.mean(relative_error),
        np.median(relative_error)
    )
```

```python
for grid_points in [51, 101, 201]:
    mean_error, median_error = (
        calculate_reconstruction_error(grid_points)
    )

    print(
        f"N = {grid_points}: "
        f"mean error = {mean_error * 100:.4f}%, "
        f"median error = {median_error * 100:.4f}%"
    )
```
