# Magnetostatic Fields and Current Sources

What magnetic field is generated by a steady current, and under what conditions can it be calculated efficiently?

## Assumptions and Scope

The note is restricted to magnetostatics.

- Current distributions are time-independent.
- Fields are evaluated in vacuum unless otherwise specified.
- Displacement current is not included.
- Magnetic materials and magnetization are not yet considered.
- Radiation and electromagnetic-wave effects are neglected.

## From Moving Charges to Steady Currents

1. Moving charges make a magnetic field.
2. Charges' continuous move can describe the current.
3. In magnetostatics, we assume the current distribution is constant over time.

$$ \frac{\partial \rho}{\partial t} = 0, \quad \frac{\partial \mathbf{J}}{\partial t} = 0$$
$$ \nabla \cdot \mathbf{J} = -\frac{\partial \rho}{\partial t}$$

in steady state current, 
$$\nabla \cdot \mathbf{J} = 0$$

Steady current does not mean that charges are stationary. It means that the current distribution is time-independent.

## Magnetic Flux Density as a Vector Field

### Definition Through the Lorentz Force

$$\mathbf{F} = q(\mathbf{E} + \mathbf{v} \times \mathbf{B})$$

When we separate the magnetic term,
$$\mathbf{F}_b = q\mathbf{v} \times \mathbf{B}$$

$\mathbf{B}$ is not a magnetic force itself; it expresses the field that describes the magnetic field. 

The magnetic force is perpendicular to both $\mathbf{v}$ and $\mathbf{B}$, and it makes the point's moving direction rather than speed. Magnetic field lines are a visualization of the direction and relative magnitude of the vector field

### The Biot-Savart Law

$$d\mathbf{B} = \frac{\mu_0}{4\pi}\frac{I d\vec{\ell'} \times (\mathbf{r}-\mathbf{r'})}{\left|\mathbf{r}-\mathbf{r'}\right| ^3}$$

$$\hat{\mathbf{R}} = (\mathbf{r}-\mathbf{r'}) $$

$\mathbf{r'}$: source point

$\mathbf{r}$: observation point

$d\ell'$: source element follows the current direction

$\hat{\mathbf{R}}$: vector that toward from source point to observation point

### Direction of the Field Contribution

Direction is depends on the right-hand screw rule

$$d\ell' \times \hat{\mathbf{R}}$$

$d\ell'$: It's toward the current flow

$\hat{\mathbf{R}}$: vector that toward from source point to observation point

$d\mathbf{B}$: Perpendicular to the plane formed by the two vectors.

$$\mathbf{B(r)} = \frac{\mu_0 I}{4\pi} \int_C\frac{d\vec{\ell'} \times (\mathbf{r}-\mathbf{r'})}{\left|\mathbf{r}-\mathbf{r'}\right| ^3}$$

### Magnetic field $\mathbf{B}$ at P due to infinite current



$$I d\ell = dz\vec{a_z}, \quad \mathbf{R} = \rho\vec{a}_\rho - z\vec{a}_z$$

$$\vec{a}_R = \frac{1}{\sqrt {\rho ^2 + z^2}} \,(\rho\vec{a}_\rho - z\vec{a}_z)$$

If we substitute this into the Biot-Savart Law's equation,
$$d\mathbf{B} = \frac{\mu _0 I}{4\pi} \frac{dz\vec{a}_z}{\rho ^2 + z^2} \times \frac{1}{\sqrt {\rho ^2 + z^2}}\,(\rho\vec{a}_\rho - z\vec{a}_z) $$
If we integrate this,
$$\mathbf{B} = \frac{\mu _0 I \rho \, \vec{a}_\phi}{4\pi} \int_{-\infty}^{\infty} \frac{dz}{(\rho ^2 + z^2)^{3/2}}$$
When we organize this using an integration table,
$$\mathbf{B} = \frac{\mu _0 I \rho \, \vec{a}_\phi}{4\pi} \frac{2}{\rho ^2}$$
We can get this relation.
$$\mathbf{B} = \frac{\mu _0 I}{2 \pi \rho}\vec{\phi}$$

The magnetic-field magnitude is directly proportional to the current
and inversely proportional to the radial distance from the wire.

$$B\propto I,\qquad B\propto \frac{1}{\rho}.$$

The field circulates around the wire in the azimuthal direction,
as determined by the right-hand rule.

### Magnetic Flux, $\Phi _B$

$$\Phi _B = \int_{S} \mathbf{B} \cdot d\mathbf{A}$$

$\mathbf{B}$: magnetic flux density, it is a local vector field

$\Phi _B$: magnetic flux, surface integral of the field

Magnetic flux density $\mathbf{B}$ is a local vector field, whereasmagnetic flux $\Phi_B$ is a scalar obtained by integrating the normal component of $\mathbf{B}$ over a surface.

### Gauss's Law for Magnetism

The Biot–Savart law determines the magnetic field generated by currents, 

$$\oint_s \mathbf{B} \cdot d\mathbf{A} = 0$$
$$\nabla \cdot \mathbf{B} = 0$$

while Gauss’s law for magnetism constrains the global structure of that field. 

The net magnetic flux through any closed surface is zero, which means that magnetic field lines have no beginning or end.

## Ampere's Circuital Law

In magnetostatics, Ampère's circuital law states that

$$\oint_C \mathbf{B}\cdot d\boldsymbol{\ell}=\mu_0 I_{\mathrm{enc}}.$$

The law is generally valid for steady currents. However, it becomes a direct and efficient method for calculating $\mathbf{B}$ only when the current distribution has sufficient symmetry.

Magnetic field with **symmetrical** current distribution, the line integral of the magnetic field vector $\mathbf{B}$ along a closed path is equal to the net current $I_\text{enc}$ enclosed by the closed path.

$$I_\text{enc} = \frac{1}{\mu _0}\oint_{C} \mathbf{B} \cdot d\vec{\ell}$$

With Stokes' theorem,
$$I_\text{enc} = \frac{1}{\mu _0}\oint_{C} (\nabla \times \mathbf{B}) \cdot d\vec{s}$$

$$I = \int \mathbf{J} \cdot d\vec{s}$$


then, we can get 
$$\nabla \times \mathbf{B} = \mu _0 \mathbf{J}$$
and,
$$\nabla \cdot \mathbf{B} = 0$$

Magnetic fields have no divergence sources, but they possess circulation around electric currents.

### Magnetic field circular loop


$$\mu _0 I_\text{enc} = \oint \mathbf{B} \cdot d\vec{\ell}$$
$$d\vec{\ell}=\rho\,d\phi\,\vec{a}_{\phi}$$
$$\mu _0 I_\text{enc} = \int_{0}^{2\pi} \mathbf{B} \cdot \rho\,d\phi\,\vec{a}_{\phi} $$
$$\mu _0 I_\text{enc} = B \int_{0}^{2\pi} \rho d\phi$$
$$B = \frac{\mu _0 I_\text{enc}}{2 \pi \rho}$$

### Tororid


when
$$ \rho _0 - a < \rho < \rho _0 + a,$$
$$\mathbf{B} = \frac{\mu _0 N I}{2 \pi \rho}\hat{\phi}$$


## Biot–Savart Law versus Ampere’s Law

|Biot–Savart Law|Ampere’s Law|
|:--------------------:|:--------------------:|
|Use direct integration|If symmetry exists, it can be calculated using a closed line integral.|
|Suitable for finite wires and current loops| Suitable for infinite wires, solenoids and toroids|
