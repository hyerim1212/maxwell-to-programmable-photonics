# Electromagnetic Waves and Dielectric Interfaces

## Physical Question

How do electromagnetic waves propagate in dielectric media, and how do Maxwell's equations determine reflection, refraction, and total internal reflection at dielectric interfaces?

## 1. Electromagnetic Waves in Homogeneous Media

### 1.1 Wave Equation
From the well-known wave formula,
$$\psi = A\sin (kx \pm \omega t + \varepsilon) $$
$$\varphi = (kx \pm \omega t + \varepsilon)$$

The following relationships can be observed.
$$\left| \frac{\partial \varphi}{\partial t} \right| = \omega , \quad \left| \frac{\partial \varphi}{\partial x} \right| = k $$
$$\left| \frac{\partial x}{\partial t} \right| = \frac{\omega}{k} = v$$

Before examining the wave equation, if we go over the complex number representation,
$$e^{i\theta} = \cos\theta + i\sin\theta$$
$$\cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2}, \quad \sin\theta = \frac{e^{i\theta} - e^{-i\theta}}{2i}$$

I regarded Maxwell's equations in `06_maxwell_equations_and_electromagnetic_induction`
$$\nabla ^2 \mathbf{E} = \epsilon  \mu  \frac{\partial ^2 \mathbf{E}}{\partial t^2}$$
$$\nabla ^2 \mathbf{B} = \epsilon  \mu  \frac{\partial ^2 \mathbf{B}}{\partial t^2}$$

and if we write the wave equation using complex number notation,
$$\psi (x, t) = \text{Re}[Ae^{i(\omega t -kx + \varepsilon)}] = A\cos(\omega t -kx + \varepsilon) $$

Let's apply the Laplacian to the wave equation.
$$\frac{\partial ^2 \psi}{\partial x ^2} + \frac{\partial ^2 \psi}{\partial y ^2} + \frac{\partial ^2 \psi}{\partial z  ^2} 
= \frac{1}{v^2}\frac{\partial ^2 \psi}{\partial t ^2}$$
$$\nabla ^2 \psi = \frac{1}{v^2}\frac{\partial ^2 \psi}{\partial t ^2}$$

Comparing the electromagnetic wave equations obtained from Maxwell’s equations with the general wave equation,

if we compare this to the wave equation, the following equation can be derived.
$$\frac{1}{v^2} = \epsilon  \mu $$
$$v = \frac{1}{\sqrt{\epsilon \mu}}$$

In a nonmagnetic dielectric,
$$\mu _r \simeq 1, \quad n = \frac{c}{v}$$
$$n \simeq \sqrt{\epsilon _r}$$

The material modifies the propagation speed of the electromagnetic wave through its constitutive response, 

which is represented by $\epsilon$ and $\mu$.

### 1.2 Harmonic Plane Waves

A plane perpendicular to $\mathbf{k}$ can be written as follows.
$$\mathbf{k} \cdot \mathbf{r} = \text{constant} $$

If we write the wave equation in terms of a plane wave,
$$\psi (\mathbf{r}) = A\sin(\mathbf{k} \cdot \mathbf{r})$$
$$\psi (\mathbf{r}) = A\cos(\mathbf{k} \cdot \mathbf{r})$$
or
$$\psi (\mathbf{r}) = Ae^{i(\mathbf{k} \cdot \mathbf{r})}$$

when $\mathbf{k} \parallel \text{x-axis}$,
$$\psi (\mathbf{r}, t) = Ae^{i(kx \, \mp \, \omega t)}$$

We can apply the results that we get from the plane wave in general.

In a nonmagnetic dielectric medium, when we use the Fourier method, 
$$\mathbf{E}_y(x,t) = E_{0, y}\cos[\omega (t - \frac{x}{v}) + \varepsilon]$$
When the direction of oscillation of the electric field lies within a plane, 

let's set the vector direction of the electric field parallel to the y-axis.
$$\mathbf{B}_z = - \int\frac{\partial \mathbf{E}_y}{\partial x} dt$$
$$\mathbf{B}_z = - \frac{E_{0, y} \omega}{v} \int \sin[\omega(t - \frac{x}{v}) + \varepsilon] dt$$
$$\mathbf{B}_z = \frac{E_{0, y}}{v}\cos[\omega (t - \frac{x}{v}) + \varepsilon] $$

Summarizing those yields the following relationship.
$$\mathbf{E}_y = v\mathbf{B}_z$$

plus,
$$\mathbf{B} = \frac{1}{\omega}\mathbf{k}\times\mathbf{E}$$

so,
$$\mathbf{E} \, \bot \, \mathbf{B} \, \bot \, \mathbf{K}$$

for magnitude,
$$B_0 = \frac{E_0}{v}$$

### 1.3 Wave Number, Phase Velocity, and Refractive Index

$$\varphi = (kx \pm \omega t + \varepsilon)$$
$$k = \frac{2\pi}{\lambda}, \quad \frac{\omega}{k} = v, \quad v = f\lambda $$
In a vacuum,
$$k_0 = \frac{\omega}{c}$$
In the medium
$$k = nk_0$$
Therefore,
$$\lambda = \frac{\lambda _0}{n}$$

When a wave passes through an interface, its frequency can be preserved, but its phase velocity and wavelength can change.
So, the change in refractive index means the change in wave vector magnitude.

## 2. Energy Transport

### 2.1 Poynting Vector
Let's define energy density $u$: Radiant energy per unit volume.
In a capacitor,
$$u_E = \frac{\frac{1}{2}CV^2}{Ad} = \frac{\frac{1}{2}\epsilon _0 (A /d)(Ed)^2}{Ad}$$
$$u_E = \frac{\epsilon _0 E^2}{2}$$

In a solenoid,
$$u_B = \frac{\frac{1}{2}L I^2}{Al} = \frac{\frac{1}{2}(\mu _0 u'^2 AI)(\frac{B}{\mu _0u'})^2}{Al}$$
$$u_B = \frac{1}{2\mu _0}B^2$$

By the relationship of $c = 1/\sqrt{\epsilon _0 \mu _0}$

$$u_E = u_B$$
$$u = u_E + u_B$$
$$u = \epsilon _0 E^2 = \frac{1}{\mu _0}B^2$$

When S is the energy transferred over time through a unit area,
$$S = \frac{uc\Delta A }{\Delta t A} = uc$$

Based on the relationship between $\mathbf{E}$ and $\mathbf{B}$ examined earlier,
$$S = \epsilon _0 E^2 C$$
$$S = \frac{1}{c\mu _0}E^2$$
$$S = \frac{1}{\mu _0}EB$$

Since it can be assumed that in the case of an isotropic medium, energy is transferred in the direction of wave propagation,
$$\mathbf{S} = \mathbf{E} \times \mathbf{H}$$

In Linear Isotropic Medium,
$$\mathbf{B} = \mu \mathbf{H}$$

so,
$$\mathbf{S} = \frac{1}{\mu}\mathbf{E} \times \mathbf{B}$$

In a vacuum,
$$\frac{1}{\mu _0} = c^2\epsilon _0$$
so,
$$\mathbf{S} = \frac{1}{\mu _0} \mathbf{E} \times \mathbf{B}$$
$$\mathbf{S} = c^2\epsilon _0 \, \mathbf{E} \times \mathbf{B}$$
We call $\mathbf{S}$ is poynting vector.

Let's write the electric and magnetic fields as plane harmonic waves.
$$\mathbf{E} = \mathbf{E}_0\cos(\mathbf{k} \cdot \mathbf{r} - \omega t), \quad \mathbf{B} = \mathbf{B}_0\cos(\mathbf{k} \cdot \mathbf{r} - \omega t)$$
$$\mathbf{S} = c^2\epsilon _0 \, \mathbf{E}_0 \times \mathbf{B}_0 \cos^2(\mathbf{k} \cdot \mathbf{r} - wt)$$

This refers to the instantaneous energy flow per unit area per unit time.

### 2.2 Time-Averaged Intensity

Since $\mathbf{S}$ vibrates at such a high speed that it is difficult to measure, an averaging process is used.

Let $\langle f(t)\rangle _T$ be the time average value of a function $f(t)$ over a measured time interval $T$.
$$\langle f(t)\rangle _T = \frac{1}{T} \int_{t-\frac{T}{2}}^{t+\frac{T}{2}} f(t') dt'$$
Let's $f(t) = e^{iwt}$
$$\langle e^{i\omega t}\rangle _T = \frac{1}{T} \int_{t-\frac{T}{2}}^{t+\frac{T}{2}} e^{i\omega t'} dt'$$
$$\langle e^{i\omega t}\rangle _T = e^{i\omega t}\left( \frac{\sin (\omega\frac{T}{2})}{\omega \frac{T}{2}} \right)$$

we write
$$\frac{\sin u}{u} = \sin c(u)$$

so we can get
$$\langle \cos \omega t \rangle _T = (\sin c(u)) \cos \omega t, \quad \langle \sin \omega t \rangle _T = (\sin c(u)) \sin \omega t$$

when $T = \tau$
$$u = \omega T/2 = \pi, \quad \sin c(u) = 0$$

Therefore,
$$\langle \cos \omega t \rangle _T = \langle \sin \omega t \rangle _T = 0$$

After a sufficiently large number of periods, the area under $f(t)$ divided by the period approaches $\frac{1}{2}$
$$\langle \cos ^2 \omega t \rangle _T = \langle \sin ^2 \omega t \rangle _T = \frac{1}{2}$$

Let's apply this to $\mathbf{S}$
$$\langle \mathbf{S} \rangle _T = c^2\epsilon _0 \, \left|\mathbf{E}_0 \times \mathbf{B}_0\right| \langle \cos^2(\mathbf{k} \cdot \mathbf{r} - wt)\rangle _T$$
when $T \gg \tau$,
$$\langle \cos^2(\mathbf{k} \cdot \mathbf{r} - wt)\rangle _T = \frac{1}{2}$$
So,
$$\langle \mathbf{S} \rangle _T = \frac{c^2\epsilon _0}{2} \, \left|\mathbf{E}_0 \times \mathbf{B}_0\right| $$
$$\langle \mathbf{S} \rangle _T = \frac{c\epsilon _0}{2} \, E_0 ^2 $$

The optical intensity is identified with the magnitude of the time-averaged Poynting vector:
$$I = \langle \mathbf{S} \rangle _T = c^2\epsilon _0 \langle \mathbf{E}^2 \rangle _T $$

## 3. Electromagnetic Response of Dielectric Matter

### 3.1 Polarization and Permittivity

When an external electric field acts on the dielectric, electric dipoles form, and these dipoles alter the internal electric field. We call that electric polarization $\mathbf{P}$.

I regarded this in the `03_conductors_currents_and_dielectrics.ipynb`. To summarize the key points,

$$(\epsilon - \epsilon _0)\mathbf{E} = \mathbf{P} $$
$$\mathbf{P} = \epsilon _0 \chi _e \mathbf{E}$$
$$\mathbf{D} = \epsilon _0 \mathbf{E} + \mathbf{P}$$
so,
$$\mathbf{D} = \epsilon _0 (1 + \chi _e)  \mathbf{E}$$
$$\epsilon _r = 1 + \chi _e$$

### 3.2 Refractive Index

The ratio of the speed of light in a vacuum to that in a medium is called the absolute refractive index.
$$n = \frac{c}{v} = \pm \,\sqrt{\frac{\epsilon \mu}{\epsilon _0 \mu _0}}$$
$$n = \pm \, \sqrt{K_E K_M}$$

Since transparent non-magnetic materials within visible light have approximately $K_M = 1.0 - 2.2 \times 10^{-5}$ values,
$$n \approx \pm \, \sqrt{K_E}$$
$$n^2 \approx K_e $$

When the medium is a non-magnetic, non-dispersive, non-absorbent medium,
$$n^2 \simeq \epsilon _r$$

The incident electromagnetic field polarizes the material, and the collective material response modifies the phase propagation of the wave.

$$\mathbf{P} = q_e x N,$$
If we introduce a virtual spring and write the equation in terms of $x$,
$$x(t) = \frac{q_e/m_e}{(\omega _0 ^2 - \omega ^2)}$$
so, 
$$\mathbf{P} = \frac{q_e ^2 N E/m_e}{(\omega _0 ^2 - \omega ^2)}$$
Bacuse of $(\epsilon - \epsilon _0)\mathbf{E} = \mathbf{P} $,
$$\epsilon = \epsilon _0 + \frac{q_e ^2 N/m_e}{(\omega _0 ^2 - \omega ^2)}$$
Because $n^2 = K_e$,
$$n(w)^2 = 1 + \frac{q_e ^2 N}{\epsilon _0 m_e}\frac{1}{(\omega _0 ^2 - \omega ^2)} $$


## 4. Reflection and Refraction

### 4.1 Phase Matching
For boundary conditions to hold at all locations and at all times on the boundary surface, the phase must be the same at all locations.
$$\omega _i = \omega _r = \omega _t$$

In addition, the tangential wave-vector component must be the same.
$$\mathbf{k}_{i, \parallel} = \mathbf{k}_{r, \parallel} = \mathbf{k}_{t, \parallel}$$
To use scalar form,
$$k_i\sin\theta _i = k_r\sin\theta _r = k_t\sin\theta _t$$

In same medium, 
$$k_i = k_r$$
$$\theta _i = \theta _r$$
and
$$k_i = n_ik_0, \quad k_r = n_rk_0$$
$$n_ik_0\theta _i = n_rk_0\theta _r$$

$$n_i\sin\theta _i = n_t\sin\theta _t$$
The tangential phase must remain continuous across the interface.

### 4.2 Fermat's Principle

$$\delta \int_{A}^{B} n(\mathbf{r}) ds = 0$$
A uniform medium exhibits a straight-line propagation pattern.

Simply put, in regions with a high refractive index, electromagnetic waves tend to travel relatively shorter distances,

while in regions with a low refractive index, they tend to travel relatively longer distances.

## 5. Electromagnetic Boundary Conditions

### 5.1 Tangential Field Continuity

$$\mathbf{E}_{i, \parallel} \, +  \,\mathbf{E}_{r, \parallel} = \mathbf{E}_{t, \parallel}$$
$$\mathbf{H}_{i, \parallel} \, + \, \mathbf{H}_{r, \parallel} = \mathbf{H}_{t, \parallel}$$

In general, 
$$\hat{\mathbf{n}} \times (\mathbf{E}_2 - \mathbf{E}_1) = 0 $$
$$\hat{\mathbf{n}} \times (\mathbf{H}_2 - \mathbf{H}_1) = \mathbf{K}_f$$

### 5.2 TE and TM Polarization

**TE / s polarization**
$$\mathbf{E} \, \bot  \, \text{plane of incidence }$$

**TM / p polarization**
$$\mathbf{B} \, \parallel \,  \text{plane of incidence }$$

### 5.3 Fresnel Coefficients

**TE**

In TE, it can be assumed as follows:
$$\hat{\mathbf{k}} \times \mathbf{E} = v \mathbf{B}, \quad \hat{\mathbf{k}} \cdot \mathbf{E} = 0$$

In the boundary plane,
$$\mathbf{E}_i \, + \, \mathbf{E}_r = \mathbf{E}_t$$

According to Ampere's law,
$$ -H_i\cos\theta _i + H_r\cos\theta _r = -H_t\cos\theta _t$$

Based on the relationship between the magnetic field and the electric field confirmed above,
$$H_i = \frac{E_i}{v_i\mu _i}, \quad H_r = \frac{E_r}{v_r\mu _r}, \quad H_t = \frac{E_t}{v_t\mu _t}$$

If we use this to simplify the above equation,
$$\frac{1}{v_i\mu _i}(E_i - E_r)\cos\theta _i = \frac{1}{v_t\mu _t}E_t\cos\theta _t$$
$$\frac{n_i}{\mu _i}(E_i - E_r)\cos\theta _i = \frac{n_t}{\mu _t}E_t\cos\theta _t$$
 
$$\left(\frac{E_{0r}}{E_{0i}}\right)_\bot = \frac{\frac{n_i}{\mu _i}\cos\theta _i - \frac{n_t}{\mu _t}\cos\theta _t}{\frac{n_i}{\mu _i}\cos\theta _i + \frac{n_t}{\mu _t}\cos\theta _t}$$

Therefore,
$$r_\bot \equiv \left(\frac{E_{0r}}{E_{0i}}\right)_\bot = \frac{\frac{n_i}{\mu _i}\cos\theta _i - \frac{n_t}{\mu _t}\cos\theta _t}{\frac{n_i}{\mu _i}\cos\theta _i + \frac{n_t}{\mu _t}\cos\theta _t}$$

$$t_\bot \equiv \left(\frac{E_{0t}}{E_{0i}}\right)_\bot = \frac{2 n_i \cos\theta _i}{n_i\cos\theta _i + n_t\cos\theta_t} $$


**TM**

If you calculate it similarly to the process above, you can obtain the following result.
$$r_\parallel = \frac{n_t \cos\theta _i - n_i \cos\theta _t}{n_i\cos\theta _t + n_t\cos\theta_i} $$

$$t_\parallel = \frac{2n_i \cos\theta _i}{n_i\cos\theta _t + n_t\cos\theta_i} $$

$r$: amplitude reflection coefficient

$t$: amplitude transmission coefficient

### 5.4 Reflectance and Transmittance

$$R = \left|r\right| ^2$$
$$T = \frac{n_t\cos\theta _t}{n_i\cos\theta _i} \left|t\right| ^2$$

$R$: reflectance

$T$: transmittance

In a non-absorbent medium,
$$R + T = 1$$

## 6. Total Internal Reflection

### 6.1 Critical Angle

When,
$$n_i > n_t$$
critical angle $\theta_c$ is,
$$\theta _t = 90\degree$$
$$\theta_c = \theta_i$$

$$\sin\theta_c = \frac{n_2}{n_1}$$

### 6.2 Evanescent Fields
The wave function for the transmitted electric field is,
$$\mathbf{E}_t = \mathbf{E}_{0t} \text{exp} i(\mathbf{k}\cdot\mathbf{r} - \omega t)$$

so,
$$\mathbf{k} \cdot \mathbf{r} = k_{tx}x + k_{ty}y$$

There's no z ingredients,
$$k_{tx} = k_{tk}\sin\theta_t, \quad k_{ty} = k_{tk}\sin\theta_y$$

If you use Snell's Law,
$$k_tcos\theta_t = \pm \, k_t \, \left(1 - \frac{\sin^2\theta_i}{n_{ti}^2} \right)^{1/2}$$

when $\sin\theta_i > n_{ti}$
$$k_{ty} = \pm \, ik_t \left(\frac{\sin^2\theta_i}{n_{ti}^2} - 1 \right)^{1/2} = \pm \, \alpha$$

Therefore, 
$$\mathbf{E}_t = \mathbf{E}_{0t}e^{ \mp \alpha y}e^{i(k_{t}x\sin\theta_i/n_{ti} - \omega t)}$$

TIR does not imply zero field in the second medium. $k_\bot$ becomes to imaginary,
$$E \approx e^{-\alpha y}$$

### 6.3 Penetration Depth

$$\alpha = \frac{2\pi n_t}{\lambda_0}\left[ \left(\frac{n_i}{n_t}\right)^2 \sin^2\theta_i -1 \right]^{1/2}$$

$\alpha$: attenustion coefficient

The strength of the evanescent electric field has a maximum value at the boundary($y = 0$) and decreases exponentially with distance from it toward the optically less dense medium. This becomes $1/e$ of the maximum value at $y = 1 / \alpha = \delta$, which is denoted as penetration depth.

## Case 1: Fresnel Reflection versus Incident Angle

### Physical Question

How do incident angle and polarization affect the reflection and transmission of an electromagnetic wave at a dielectric interface?

### Model and Conditions

Consider a monochromatic plane electromagnetic wave incident from a higher-index dielectric medium onto a lower-index dielectric medium.

$$n_1 = 1.50, \qquad n_2 = 1.00$$

Assume both media are:

- linear
- isotropic
- homogeneous
- nonmagnetic
- lossless

The incident angle is varied over

$$0^\circ \leq \theta_i < \theta_c$$

The transmission angle is determined from Snell's law,

$$n_1 \sin\theta_i = n_2 \sin\theta_t$$

### Tasks

#### Task A — Fresnel Coefficients

Calculate the Fresnel amplitude reflection and transmission coefficients for both TE and TM polarization:

$$r_s(\theta_i), \qquad t_s(\theta_i)$$

$$r_p(\theta_i), \qquad t_p(\theta_i)$$

#### Task B — Reflectance and Transmittance

Calculate the reflectance and transmittance,

$$R = |r|^2$$

$$T = \frac{n_2\cos\theta_t}{n_1\cos\theta_i}|t|^2$$

Plot the reflectance for TE and TM polarization as functions of incident angle:

$$R_s(\theta_i), \qquad R_p(\theta_i)$$

#### Task C — Energy Conservation Verification

For the lossless interface, verify that

$$R + T = 1.$$

Define an energy-conservation residual,

$$\epsilon_{\mathrm{energy}}=|R+T-1|$$

and calculate the maximum residual over the valid incident-angle range.

Also verify that at normal incidence,

$$R_s = R_p.$$

#### Task D — Brewster Angle

For TM polarization, numerically determine the incident angle at which the reflectance becomes minimum.

Compare the numerical result with the theoretical Brewster angle,

$$\theta_B = \tan^{-1}\left(\frac{n_2}{n_1}\right)$$

Evaluate the absolute error,

$$\Delta\theta_B = |\theta_{B,\mathrm{num}}-\theta_{B,\mathrm{theory}}|$$

#### Task E — Total Internal Reflection

Calculate the critical angle,

$$\theta_c=\sin^{-1}\left(\frac{n_2}{n_1}\right)$$

Examine the behavior of the reflectance for


$$\theta_i > \theta_c.$$

### Physical Interpretation

Discuss the following questions:

1. Why do TE and TM polarizations have the same reflectance at normal incidence?
2. Why does a Brewster angle occur only for TM polarization?
3. How does the reflectance change as the incident angle approaches and exceeds the critical angle?
4. Does \(R=1\) imply that the electromagnetic field in the second medium is zero?

```python
n_i = 1.50
n_t = 1.00

#E
theta_c = critical_angle(n_i, n_t)

incident_angles = np.linspace(0.0, theta_c - 1e-6,200)
incident_angles_deg = np.degrees(incident_angles)

transmitted_angles = snell_angle(n_i,n_t,incident_angles)

TE = fresnel_coefficients(n_i, n_t, incident_angles, "TE")
TM = fresnel_coefficients(n_i, n_t, incident_angles, "TM")

r_s = TE["r_s"]
t_s = TE["t_s"]

r_p = TM["r_p"]
t_p = TM["t_p"]

R_TE = reflectance(r_s)
R_TM = reflectance(r_p)

T_TE = transmittance(t_s, n_i, n_t, incident_angles, transmitted_angles)

T_TM = transmittance(t_p, n_i, n_t, incident_angles, transmitted_angles)

verify_TE = power_conservation_error(R_TE, T_TE)
verify_TM = power_conservation_error(R_TM, T_TM)

energy_conservation_residual_TE = np.abs(R_TE + T_TE - 1)
energy_conservation_residual_TM = np.abs(R_TM + T_TM - 1)

theta_B_numerical = incident_angles[np.argmin(R_TM)]
theta_B_theory = np.arctan(n_t / n_i)

delta_theta_B = abs(theta_B_numerical - theta_B_theory)
```

```python
print(f"Brewster angle (numerical): {np.degrees(theta_B_numerical):.3f} deg")
print(f"Brewster angle (theory): {np.degrees(theta_B_theory):.3f} deg")
print(f"Delta theta_B: {np.degrees(delta_theta_B):.3f} deg")
print(f"Critical angle: {np.degrees(theta_c):.3f} deg")

print(
    f"Maximum TE conservation error: "
    f"{np.max(energy_conservation_residual_TE):.3e}"
)

print(
    f"Maximum TM conservation error: "
    f"{np.max(energy_conservation_residual_TM):.3e}"
)
```

```python
plt.figure(figsize=(6,4))

plt.plot(
    incident_angles_deg,
    R_TE,
    label = r"$R_{TE}(\theta_i)$",
    color = COLORS["numerical"]
)

plt.plot(
    incident_angles_deg,
    R_TM,
    label = r"$R_{TM}(\theta_i)$",
    color = COLORS["numerical_alt"]
)

plt.xlabel("Incident Angle")
plt.ylabel("Reflectance")
plt.title("Reflectance versus Incident Angle")

plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()
```

```python
energy_conservation_residual_TE = abs(R_TE + T_TE - 1)
energy_conservation_residual_TM = abs(R_TM + T_TM - 1)

plt.figure(figsize=(6,4))

plt.scatter(
    incident_angles_deg,
    energy_conservation_residual_TE,
    label = r"$Residual_{TE}$",
    color=COLORS["error"],
    marker="o",
    alpha=0.8
)

plt.scatter(
    incident_angles_deg,
    energy_conservation_residual_TM,
    label = r"$Residual_{TM}$",
    color = COLORS["error_alt"],
    marker="x",
    alpha=0.8
)

plt.xlabel("Incident Angle")
plt.ylabel("Power Conservation Error")
plt.title("Power Conservation Error versus Incident Angle")

plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()
```

## Case 2: Evanescent Penetration under Total Internal Reflection

### Physical Question

How far does an evanescent electromagnetic field penetrate into a lower-index dielectric medium during total internal reflection, and what determines the penetration depth?

### Model and Conditions

Use the same dielectric interface as in Case 1:

$$n_1 = 1.50, \qquad n_2 = 1.00$$

Assume both media are:

- linear
- isotropic
- homogeneous
- nonmagnetic
- lossless

Use a vacuum wavelength of

$$\lambda_0 = 1.55\ \mu\mathrm{m}$$

The wave is incident from medium 1 at an angle satisfying

$$\theta_i > \theta_c$$

where

$$\theta_c = \sin^{-1}\left(\frac{n_2}{n_1}\right)$$

### Tasks

#### Task A — Evanescent Decay Constant

Define the vacuum wave number,

$$k_0=\frac{2\pi}{\lambda_0}$$

Under total internal reflection, calculate the evanescent decay constant in medium 2,

$$\alpha=k_0\sqrt{n_1^2\sin^2\theta_i-n_2^2}$$

Model the normalized field amplitude in the second medium as

$$\frac{|E(y)|}{|E(0)|}=e^{-\alpha y}$$

#### Task B — Evanescent Field Profiles

Calculate and compare the normalized field profiles for several incident angles above the critical angle.

Use, for example,

$$\theta_i=\theta_c+1^\circ,\qquad\theta_c+5^\circ,\qquad\theta_c+15^\circ$$

Plot

$$\frac{|E(y)|}{|E(0)|}$$

as a function of distance from the interface.

#### Task C — Penetration Depth

Define the amplitude penetration depth as

$$\delta=\frac{1}{\alpha}$$

Therefore,

$$\delta=\frac{\lambda_0}{2\pi\sqrt{n_1^2\sin^2\theta_i-n_2^2}}$$

Vary the incident angle over

$$\theta_c < \theta_i < 90^\circ$$

and calculate

$$\delta(\theta_i)$$

Plot the penetration depth as a function of incident angle.

#### Task D — Limiting Behavior

Investigate the behavior of the penetration depth as

$$\theta_i \rightarrow \theta_c^+$$

and as

$$\theta_i \rightarrow 90^\circ$$

Explain physically why the evanescent field penetrates farther into the second medium when the incident angle is only slightly above the critical angle.

#### Task E — Wavelength Dependence

Choose a fixed incident angle above the critical angle and calculate the penetration depth for several vacuum wavelengths, for example,

$$\lambda_0=1.0,\ 1.3,\ 1.55,\ 2.0\ \mu\mathrm{m}$$

Verify numerically whether

$$\delta \propto \lambda_0$$

### Physical Interpretation

Discuss the following questions:

1. Why does the transmitted field not vanish even though no net propagating power is transmitted into the second medium under total internal reflection?
2. Why does the penetration depth increase as the incident angle approaches the critical angle?
3. How does wavelength affect the spatial extent of the evanescent field?
4. How does evanescent penetration help explain field confinement in a dielectric waveguide?
5. What would be expected if another high-index dielectric structure were placed within the evanescent-field region?

```python
wavelength_vacuum = 1.55e-6

incident_angles_2 = np.radians([45, 50, 60])

alpha  = evanescent_decay_constant(wavelength_vacuum, n_i, n_t, incident_angles_2)
angles = theta_c + np.radians([1, 5, 15])

penetrate_depth = []

incident_angles_c = np.linspace(theta_c + np.radians(0.1), np.pi /2, 200)

for theta in incident_angles_c:
    alpha  = evanescent_decay_constant(wavelength_vacuum, n_i, n_t, theta)
    delta = 1 / alpha
    penetrate_depth.append(delta)

penetrate_depth = np.asarray(penetrate_depth)

theta_offsets_deg = np.array([5.0, 1.0, 0.5, 0.1, 0.01])

theta_near_critical = (
    theta_c + np.radians(theta_offsets_deg)
)

alpha_near_critical = evanescent_decay_constant(
    wavelength_vacuum,
    n_i,
    n_t,
    theta_near_critical
)

penetration_depth_near_critical = 1 / alpha_near_critical

theta_near_90_deg = np.array([80.0, 85.0, 88.0, 89.0, 89.9])

theta_near_90 = np.radians(theta_near_90_deg)

alpha_near_90 = evanescent_decay_constant(
    wavelength_vacuum,
    n_i,
    n_t,
    theta_near_90
)

penetration_depth_near_90 = 1 / alpha_near_90

k_0 = 2 * np.pi / wavelength_vacuum

alpha_limit_90 = (
    k_0
    * np.sqrt(n_i**2 - n_t**2)
)

penetration_depth_limit_90 = 1 / alpha_limit_90

l = np.array([1.0e-6, 1.3e-6, 1.55e-6, 2.0e-6])
alpha  = evanescent_decay_constant(l, n_i, n_t, np.pi/4)
delta = 1 / alpha

ratio = delta / l
```

```python
print(
    f"alpha: {alpha}")

print("Near-critical behavior")
print()

for offset, alpha_value, delta_value in zip(
    theta_offsets_deg,
    alpha_near_critical,
    penetration_depth_near_critical):
    print(
        f"theta_i - theta_c = {offset:6.2f} deg | "
        f"alpha = {alpha_value:.3e} 1/m | "
        f"delta = {delta_value * 1e6:.3f} um"
    )

print()
print("theta_i -> 90 deg behavior")
print()

for angle_deg, alpha_value, delta_value in zip(
    theta_near_90_deg,
    alpha_near_90,
    penetration_depth_near_90
):
    print(
        f"theta_i = {angle_deg:5.1f} deg | "
        f"alpha = {alpha_value:.3e} 1/m | "
        f"delta = {delta_value * 1e6:.3f} um"
    )

print()
print(
    f"Analytical alpha limit: "
    f"{alpha_limit_90:.3e} 1/m"
)

print(
    f"Analytical penetration-depth limit: "
    f"{penetration_depth_limit_90 * 1e6:.3f} um"
)

print(f"ratio: {ratio}")
```

### Physical Interpretation

Q1. Why does the transmitted field not vanish even though no net propagating power is transmitted into the second medium under total internal reflection?

A1. 전자기적 경계 조건에 따라 전기장과 자기장이 경계면을 가로질러 연속적이어야 하기 때문이다. 전반사 상황에서 이러한 조건을 만족시키기 위해서는 $\text{time-averaged normal Poynting component} = 0$이어도, 0이 아닌 다른 장이 형성되어야 하며, 이를 경계파라고 한다.

Q2. Why does the penetration depth increase as the incident angle approaches the critical angle?

A2. penetration depth는 경계면에서의 전기장 강도의 $1/e$가 되는 지점이다. 
$$\alpha = \frac{2\pi n_t}{\lambda_0}\left[ \left(\frac{n_i}{n_t}\right)^2 \sin^2\theta_i -1 \right]^{1/2}$$
인 감쇠계수의 역수로 정의된다.

입사각이 임계각에 가까워지면 굴절각이 90도에 가까워진다. 경계면에서의 정상파 크기는 에바네센트파의 크기와 일치하고, 이는 감쇠계수가 줄어들어 penetration depth가 커짐을 의미한다.

Q3. How does wavelength affect the spatial extent of the evanescent field?

A3. 파장의 증가는 에바네센트장의 감소를 야기한다.

$$\alpha \propto \frac{1}{\lambda _0}, \quad \delta \propto \lambda _0$$

Q4. How does evanescent penetration help explain field confinement in a dielectric waveguide?

A4. 소멸파는 유전체 도파관의 외부인 클래딩에서 지수적으로 감소하고, 이러한 감소로 인해 대부분의 광에너지가 코어 내부에 위치하게 된다. 이는 빛의 에너지 손실 없이 광섬유 혹은 채널을 따라 이동할 수 있게 한다.

Q5. What would be expected if another high-index dielectric structure were placed within the evanescent-field region?

A5. 에바네센트장이 존재하는 영역 중간에 고굴절률 유전체가 위치한다면 전반사 조건을 만족시키지 못한다. 따라서 경계파 혹은 에바네센트파가 유전체에 의해 반사되어 중첩 혹은 감쇠되거나, 기존의 경계면에 영향을 미쳐 기존의 경계 조건을 변화시킬 수 있을 것 같다.

---

에바네센트파는 입사각이 임계각보다 커질때 감쇠 상수가 증가하여 빠르게 감소한다. 임계각 근처에서 횡방향 감쇠 상수는 0에 가까워지고 침투 깊이가 커진다. 에바네센트파는 보통 수 파장 정도의 거리를 벗어나면 거의 무시되어 파장이 길수록 침투 깊이는 깊어진다. 

The evanescent field decays more rapidly as the incident angle increases above the critical angle because the decay constant increases. Near the critical angle, the transverse decay constant approaches zero and the penetration depth becomes large, illustrating weak confinement. Longer wavelengths also produce larger penetration depths.
