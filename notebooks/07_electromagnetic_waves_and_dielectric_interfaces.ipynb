# Electromagnetic Waves and Dielectric Interfaces

## Physical Question

How do electromagnetic waves propagate in dielectric media, and how do Maxwell's equations determine reflection, refraction, and total internal reflection at dielectric interfaces?

## 1. Electromagnetic Waves in Homogeneous Media

### 1.1 Wave Equation
From the well-known wave formula,
$$\psi = A\sin (kx \pm \omega t + \varepsilon) $$
$$\varphi = (kx \pm \omega t + \varepsilon)$$

The following relationships can be observed.
$$\left| \frac{\partial \varphi}{\partial t} \right| = \omega , \quad \left| \frac{\partial \varphi}{\partial x} \right| = k $$
$$\left| \frac{\partial x}{\partial t} \right| = \frac{\omega}{k} = v$$

Before examining the wave equation, if we go over the complex number representation,
$$e^{i\theta} = \cos\theta + i\sin\theta$$
$$\cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2}, \quad \sin\theta = \frac{e^{i\theta} - e^{-i\theta}}{2i}$$

If we write the wave equation using complex number notation,
$$\psi (x, t) = \text{Re}[Ae^{i(\omega t -kx + \varepsilon)}] = A\cos(\omega t -kx + \varepsilon) $$

Let's apply the Laplacian to the wave equation.
$$\frac{\partial ^2 \psi}{\partial x ^2} + \frac{\partial ^2 \psi}{\partial y ^2} + \frac{\partial ^2 \psi}{\partial z  ^2} 
= \frac{1}{v^2}\frac{\partial ^2 \psi}{\partial t ^2}$$
$$\nabla ^2 \psi = \frac{1}{v^2}\frac{\partial ^2 \psi}{\partial t ^2}$$

Since equations for physical quantities given as functions of time and space can be explained by wave phenomena,
$$\nabla ^2 \mathbf{E} = \epsilon  \mu  \frac{\partial ^2 \mathbf{E}}{\partial t^2}$$
$$\nabla ^2 \mathbf{B} = \epsilon  \mu  \frac{\partial ^2 \mathbf{B}}{\partial t^2}$$

Therefore, the following equation can be derived.
$$\frac{1}{v^2} = \epsilon  \mu $$
$$v = \pm \frac{1}{\sqrt{\epsilon \mu}}$$

In nonmagnetic dielectirc,
$$\mu _r \simeq 1, \quad n = \frac{c}{v}$$
$$n \simeq \sqrt{\epsilon _r}$$

The material modifies the propagation speed of the electromagnetic wave through its constitutive response, 

which is represented by $\epsilon$ and $\mu$.

### 1.2 Harmonic Plane Waves

A plane perpendicular to $\mathbf{k}$ can be written as follows.
$$\mathbf{k} \cdot \mathbf{r} = \text{constant} $$

If we write the wave equation in terms of a plane wave,
$$\psi (\mathbf{r}) = A\sin(\mathbf{k} \cdot \mathbf{r})$$
$$\psi (\mathbf{r}) = A\cos(\mathbf{k} \cdot \mathbf{r})$$
or
$$\psi (\mathbf{r}) = Ae^{i(\mathbf{k} \cdot \mathbf{r})}$$

when $\mathbf{k} \parallel \text{x-axis}$,
$$\psi (\mathbf{r}, t) = Ae^{i(kx \, \mp \, \omega t)}$$

We can apply the results that we get from the plane wave in general.

In a nonmagnetic dielectric medium, when we use the Fourier method, 
$$\mathbf{E}_y(x,t) = E_{0, y}\cos[\omega (t - \frac{x}{v}) + \varepsilon]$$
When the direction of oscillation of the electric field lies within a plane, 

let's set the vector direction of the electric field parallel to the y-axis.
$$\mathbf{B}_z = - \int\frac{\partial \mathbf{E}_y}{\partial x} dt$$
$$\mathbf{B}_z = - \frac{E_{0, y} \omega}{v} \int \sin[\omega(t - \frac{x}{v}) + \varepsilon] dt$$
$$\mathbf{B}_z = \frac{E_{0, y}}{v}\cos[\omega (t - \frac{x}{v}) + \varepsilon] $$

Summarizing those yields the following relationship.
$$\mathbf{E}_y = v\mathbf{B}_z$$
### 1.3 Wave Number, Phase Velocity, and Refractive Index

$$\varphi = (kx \pm \omega t + \varepsilon)$$
$$k = \frac{2\pi}{\lambda}, \quad \frac{\omega}{k} = v, \quad v = f\lambda $$
In a vacuum,
$$k_0 = \frac{\omega}{c}$$
In the medium
$$k = nk_0$$
Therefore,
$$\lambda = \frac{\lambda _0}{n}$$

When a wave passes through an interface, its frequency can be preserved, but its phase velocity and wavelength can change.
So, the change in refractive index means the change in wave vector magnitude.

## 2. Energy Transport

### 2.1 Poynting Vector
Let's define energy density $u$: Radiant energy per unit volume.
In a capacitor,
$$u_E = \frac{\frac{1}{2}CV^2}{Ad} = \frac{\frac{1}{2}\epsilon _0 A /d(Ed)^2}{Ad}$$
$$u_E = \frac{\epsilon _0 E^2}{2}$$
In a solenoid,
$$u_B = \frac{\frac{1}{2}L I^2}{Al} = \frac{\frac{1}{2}(\mu _0 u'^2 AI)(\frac{B}{\mu _0u'})^2}{Al}$$
$$u_B = \frac{l}{2\mu _0}B^2$$

By the relationship of $c = 1/\sqrt{\epsilon _0 \mu _0}$

$$u_E = u_B$$
$$u = u_E + u_B$$
$$u = \epsilon _0 E^2 = \frac{l}{\mu _0}B^2$$

When S is the energy transferred over time through a unit area,
$$S = \frac{uc\Delta A }{\Delta t A} = uc$$
Based on the relationship between $\mathbf{E}$ and $\mathbf{B}$ examined earlier,
$$S = \epsilon _0 E^2 C$$
$$S = \frac{l}{c\mu _0}E^2$$
$$S = \frac{l}{\mu _0}EB$$
Since it can be assumed that in the case of an isotropic medium, energy is transferred in the direction of wave propagation,
$$\mathbf{S} = \frac{l}{\mu _0} \mathbf{E} \times \mathbf{B}$$
$$\mathbf{S} = c^2\epsilon _0 \, \mathbf{E} \times \mathbf{B}$$
We call $\mathbf{S}$ is poynting vector.

Let's write the electric and magnetic fields as plane harmonic waves.
$$\mathbf{E} = \mathbf{E}_0\cos(\mathbf{k} \cdot \mathbf{r} - \omega t), \quad \mathbf{B} = \mathbf{B}_0\cos(\mathbf{k} \cdot \mathbf{r} - \omega t)$$
$$\mathbf{S} = c^2\epsilon _0 \, \mathbf{E}_0 \times \mathbf{B}_0 \cos^2(\mathbf{k} \cdot \mathbf{r} - wt)$$

This refers to the instantaneous energy flow per unit area per unit time.

### 2.2 Time-Averaged Intensity

Since $\mathbf{S}$ vibrates at such a high speed that it is difficult to measure, an averaging process is used.

Let $\langle f(t)\rangle _T$ be the time average value of a function $f(t)$ over a measured time interval $T$.
$$\langle f(t)\rangle _T = \frac{1}{T} \int_{t-\frac{T}{2}}^{t+\frac{T}{2}} f{t} dt$$
Let's $f(t) = e^{iwt}$
$$\langle e^{i\omega t}\rangle _T = \frac{1}{T} \int_{t-\frac{T}{2}}^{t+\frac{T}{2}} e^{i\omega t} dt$$
$$\langle e^{i\omega t}\rangle _T = \frac{1}{i\omega t}e^{i\omega t}(e^{i\omega tT/2}-e^{-iwT/2})$$
$$\langle e^{i\omega t}\rangle _T = \left( \frac{\sin (\omega\frac{T}{2})}{\omega \frac{T}{2}} \right)$$
we write
$$\frac{\sin u}{u} = \sin cu$$
so we can get
$$\langle \cos \omega t \rangle _T = (\sin cu) \cos \omega t, \quad \langle \sin \omega t \rangle _T = (\sin cu) \sin \omega t$$
when $T = \tau$
$$u = \omega T/2 = \pi, \quad \sin cu = 0$$
Therefore,
$$\langle \cos \omega t \rangle _T = \langle \sin \omega t \rangle _T = 0$$

After a sufficiently large number of periods, the area under $f(t)$ divided by the period approaches $\frac{1}{2}$
$$\langle \cos ^2 \omega t \rangle _T = \langle \sin ^2 \omega t \rangle _T = \frac{1}{2}$$

Let's apply this to $\mathbf{S}$
$$\langle \mathbf{S} \rangle _T = c^2\epsilon _0 \, \left|\mathbf{E}_0 \times \mathbf{B}_0\right| \langle \cos^2(\mathbf{k} \cdot \mathbf{r} - wt)\rangle _T$$
when $T \gg \tau$,
$$\langle \cos^2(\mathbf{k} \cdot \mathbf{r} - wt)\rangle _T = \frac{1}{2}$$
So,
$$\langle \mathbf{S} \rangle _T = \frac{c^2\epsilon _0}{2} \, \left|\mathbf{E}_0 \times \mathbf{B}_0\right| $$
$$\langle \mathbf{S} \rangle _T = \frac{c\epsilon _0}{2} \, E_0 ^2 $$

The radiant intensity $I$, used when referring to the total amount of light illuminating a surface, can be written as follows:
$$I = \langle \mathbf{S} \rangle _T = c^2\epsilon _0 \langle \mathbf{E}^2 \rangle _T $$

## 3. Electromagnetic Response of Dielectric Matter

### 3.1 Polarization and Permittivity

When an external electric field acts on the dielectric, electric dipoles form, and these dipoles alter the internal electric field. We call that electric polarization $\mathbf{P}$.

I regard this in the `03_conductors_currents_and_dielectrics.ipynb`. To summarize the key points,

$$(\epsilon - \epsilon _0)\mathbf{E} = \mathbf{P} $$
$$\mathbf{P} = \epsilon _0 \chi _e \mathbf{E}$$
$$\mathbf{D} = \epsilon _0 \mathbf{E} + \mathbf{P}$$
so,
$$\mathbf{D} = \epsilon _0 (1 + \chi _e)  \mathbf{E}$$
$$\epsilon _r = 1 + \chi _e$$

### 3.2 Refractive Index

The ratio of the speed of light in a vacuum to that in a medium is called the absolute refractive index.
$$n = \frac{c}{v} = \pm \,\sqrt{\frac{\epsilon \mu}{\epsilon _0 \mu _0}}$$
$$n = \pm \, \sqrt{K_E K_M}$$

Since transparent non-magnetic materials within visible light have approximately $K_M = 1.0 - 2.2 \times 10^{-5}$ values,
$$n \approx \pm \, \sqrt{K_E}$$
$$n^2 \approx K_e $$

When the medium is a non-magnetic, non-dispersive, non-absorbent medium,
$$n^2 \simeq \epsilon _r$$

The incident electromagnetic field polarizes the material, and the collective material response modifies the phase propagation of the wave.

$$\mathbf{P} = q_e x N,$$
If we introduce a virtual spring and write the equation in terms of $x$,
$$x(t) = \frac{q_e/m_e}{(\omega _0 ^2 - \omega ^2)}$$
so, 
$$\mathbf{P} = \frac{q_e ^2 N E/m_e}{(\omega _0 ^2 - \omega ^2)}$$
Bacuse of $(\epsilon - \epsilon _0)\mathbf{E} = \mathbf{P} $,
$$\epsilon = \epsilon _0 + \frac{q_e ^2 N/m_e}{(\omega _0 ^2 - \omega ^2)}$$
Because $n^2 = K_e$,
$$n(w)^2 = 1 + \frac{q_e ^2 N}{\epsilon _0 m_e}\frac{1}{(\omega _0 ^2 - \omega ^2)} $$

## 4. Reflection and Refraction

### 4.1 Phase Matching
For boundary conditions to hold at all locations and at all times on the boundary surface, the phase must be the same at all locations.
$$\omega _i = \omega _r = \omega _t$$

In addition, the tangential wave-vector component must be the same.
$$\mathbf{k}_{i, \parallel} = \mathbf{k}_{r, \parallel} = \mathbf{k}_{t, \parallel}$$
To use scalar form,
$$k_i\sin\theta _i = k_r\sin\theta _r = k_t\sin\theta _t$$

In same medium, 
$$k_i = k_r$$
$$\theta _i = \theta _r$$
and
$$k_i = n_ik_0, \quad k_r = n_rk_0$$
$$n_ik_0\theta _i = n_rk_0\theta _r$$

$$n_i\sin\theta _i = n_t\sin\theta _t$$

### 4.2 Snell's Law

$$n_i\sin\theta _i = n_t\sin\theta _t$$
The tangential phase must remain continuous across the interface.

### 4.3 Fermat's Principle

$$\delta \int_{A}^{B} n(\mathbf{r}) ds = 0$$
A uniform medium exhibits a straight-line propagation pattern.

Simply put, in regions with a high refractive index, electromagnetic waves tend to travel relatively shorter distances,

while in regions with a low refractive index, they tend to travel relatively longer distances.

## 5. Electromagnetic Boundary Conditions

### 5.1 Tangential Field Continuity

$$\mathbf{E}_{i, \parallel} = \mathbf{E}_{r, \parallel}$$
$$\mathbf{H}_{i, \parallel} = \mathbf{H}_{r, \parallel}$$

### 5.2 TE and TM Polarization

**TE / s polarization**
$$\mathbf{E} \, \bot  \, \text{plane of incidence }$$

**TM / p polarization**
$$\mathbf{B} \, \parallel \,  \text{plane of incidence }$$

### 5.3 Fresnel Coefficients

**TE**

In TE, it can be assumed as follows:
$$\hat{\mathbf{k}} \times \mathbf{E} = v \mathbf{B}, \quad \hat{\mathbf{k}} \cdot \mathbf{E} = 0$$

In the boundary plane,
$$\mathbf{E}_i + \mathbf{E}_r = \mathbf{E}_t$$

According to Ampere's law,
$$ -H_i\cos\theta _i + H_r\cos\theta _r = -H_t\cos\theta _t$$

Based on the relationship between the magnetic field and the electric field confirmed above,
$$H_i = \frac{E_i}{v_i\mu _i}, \quad H_r = \frac{E_r}{v_r\mu _r}, \quad H_t = \frac{E_t}{v_t\mu _t}$$

If we use this to simplify the above equation,
$$\frac{1}{v_i\mu _i}(E_i - E_r)\cos\theta _i = \frac{1}{v_t\mu _t}E_t\cos\theta _t$$
$$\frac{n_i}{\mu _i}(E_i - E_r)\cos\theta _i = \frac{n_t}{\mu _t}E_t\cos\theta _t$$
 
$$\left(\frac{E_{0r}}{E_{0i}}\right)_\bot = \frac{\frac{n_i}{\mu _i}\cos\theta _i - \frac{n_t}{\mu _t}\cos\theta _t}{\frac{n_i}{\mu _i}\cos\theta _i + \frac{n_t}{\mu _t}\cos\theta _t}$$

Therefore,
$$r_\bot \equiv \left(\frac{E_{0r}}{E_{0i}}\right)_\bot = \frac{\frac{n_i}{\mu _i}\cos\theta _i - \frac{n_t}{\mu _t}\cos\theta _t}{\frac{n_i}{\mu _i}\cos\theta _i + \frac{n_t}{\mu _t}\cos\theta _t}$$

$$t_\bot \equiv \left(\frac{E_{0t}}{E_{0i}}\right)_\bot = \frac{2 n_i \cos\theta _i}{n_i\cos\theta _i + n_t\cos\theta_t} $$


**TM**

If you calculate it similarly to the process above, you can obtain the following result.
$$r_\parallel = \frac{n_t \cos\theta _i - n_i \cos\theta _t}{n_i\cos\theta _t + n_t\cos\theta_i} $$

$$t_\parallel = \frac{2n_i \cos\theta _i}{n_i\cos\theta _t + n_t\cos\theta_i} $$

$r$: amplitude reflection coefficient

$t$: amplitude transmission coefficient

### 5.4 Reflectance and Transmittance

$$R = \left|r\right| ^2$$
$$T = \frac{n_t\cos\theta _t}{n_i\cos\theta _i} \left|t\right| ^2$$

$R$: reflectance

$T$: transmittance

In a non-absorbent medium,
$$R + T = 1$$

## 6. Total Internal Reflection

### 6.1 Critical Angle

When,
$$n_i > n_t$$
critical angle $\theta_c$ is,
$$\theta _t = 90\degree$$
$$\theta_c = \theta_i$$

$$\sin\theta_c = \frac{n_2}{n_1}$$

### 6.2 Evanescent Fields
The wave function for the transmitted electric field is,
$$\mathbf{E}_t = \mathbf{E}_{0t} \text{exp} i(\mathbf{k}\cdot\mathbf{r} - \omega t)$$

so,
$$\mathbf{k} \cdot \mathbf{r} = k_{tx}x + k_{ty}y$$

There's no z ingredients,
$$k_{tx} = k_{tk}\sin\theta_t, \quad k_{ty} = k_{tk}\sin\theta_y$$

If you use Snell's Law,
$$k_tcos\theta_t = \pm \, k_t \, \left(1 - \frac{\sin^2\theta_i}{n_{ti}^2} \right)^{1/2}$$

when $\sin\theta_i > n_{ti}$
$$k_{ty} = \pm \, ik_t \left(\frac{\sin^2\theta_i}{n_{ti}^2} - 1 \right)^{1/2} = \pm \, \beta$$

Therefore, 
$$\mathbf{E}_t = \mathbf{E}_{0t}e^{ \mp \beta y}e^{i(k_{t}x\sin\theta_i/n_{ti} - \omega t)}$$

### 6.3 Penetration Depth

$$\beta = \frac{2\pi n_t}{\lambda_0}\left[ \left(\frac{n_i}{n_t})\right)^2 sin^2\theta_i -1 \right]^{1/2}$$

$\beta$: attenustion coefficient

The strength of the evanescent electric field has a maximum value at the boundary($y = 0$) and decreases exponentially with distance from it toward the optically less dense medium. This becomes $1/e$ of the maximum value at $y = 1 / \beta = \delta$, which is denoted as penetration depth.
